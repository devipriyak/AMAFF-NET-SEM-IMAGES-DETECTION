```python

# =====================================================================
# AMAFF-Net_SE: SCI-Quality Proposed Architecture Diagram
#
# Architecture:
# Input Image
#    |
# Model-Specific Preprocessing
#    |
# Three Parallel CNN Branches:
# Xception + SE | DenseNet201 + SE | MobileNetV2 + SE
#    |
# Global Average Pooling + Batch Normalization
#    |
# 512-D Feature Projection (each branch)
#    |
# Feature Concatenation (1536-D)
#    |
# Adaptive Branch-Weight Generation
#    |
# Weighted Feature Fusion (512-D)
#    |
# Classification
#
# Grad-CAM is represented separately for interpretability.
# =====================================================================

# STEP 1: Install Graphviz in Google Colab
!apt-get -qq install graphviz
!pip install -q graphviz

# STEP 2: Import libraries
from graphviz import Digraph
from IPython.display import display, Image

# STEP 3: Create the architecture diagram
dot = Digraph(
    "AMAFFNet_SE",
    format="png"
)

# STEP 4: Set publication-quality graph properties
dot.attr(
    rankdir="TB",
    splines="ortho",
    nodesep="0.65",
    ranksep="0.65",
    pad="0.35",
    bgcolor="white",
    dpi="600",
    compound="true",
    newrank="true",
    labelloc="t",
    label="Proposed Adaptive Multi-scale Attention Feature Fusion Network (AMAFF-Net_SE)",
    fontsize="22",
    fontname="Arial Bold",
    fontcolor="#172B4D"
)

# General node appearance
dot.attr(
    "node",
    shape="box",
    style="rounded,filled",
    fontname="Arial",
    fontsize="11",
    margin="0.20,0.13",
    penwidth="1.4",
    color="#53657A",
    fontcolor="#172B4D"
)

# General arrow appearance
dot.attr(
    "edge",
    color="#526579",
    penwidth="1.4",
    arrowsize="0.8",
    fontname="Arial",
    fontsize="9",
    fontcolor="#34495E"
)

# STEP 5: Input image and common preprocessing
dot.node(
    "input",
    "Input Microscopic RGB Image",
    fillcolor="#EAF2F8",
    color="#2874A6",
    penwidth="1.8"
)

dot.node(
    "preprocessing",
    "Image Preprocessing",
    fillcolor="#E8F5E9",
    color="#388E3C"
)

dot.edge("input", "preprocessing")

# STEP 6: Create three parallel feature extraction branches
# Each branch has its own model-specific input resolution.

branch_data = [
    {
        "id": "x",
        "name": "Xception Branch",
        "resolution": "299 x 299",
        "cnn": "Xception CNN",
        "color": "#DCEAF7",
        "border": "#2874A6"
    },
    {
        "id": "d",
        "name": "DenseNet201 Branch",
        "resolution": "224 x 224",
        "cnn": "DenseNet201 CNN",
        "color": "#E2F0D9",
        "border": "#548235"
    },
    {
        "id": "m",
        "name": "MobileNetV2 Branch",
        "resolution": "160 x 160",
        "cnn": "MobileNetV2 CNN",
        "color": "#FCE4D6",
        "border": "#C65911"
    }
]

# Store the final feature node identifiers
feature_nodes = []

for branch in branch_data:

    p = branch["id"]
    branch_name = branch["name"]
    resolution = branch["resolution"]
    cnn_name = branch["cnn"]
    branch_color = branch["color"]
    branch_border = branch["border"]

    # Create a separate visual cluster for each CNN branch
    with dot.subgraph(name=f"cluster_{p}") as c:

        c.attr(
            label=branch_name,
            labelloc="t",
            labeljust="c",
            fontsize="13",
            fontname="Arial Bold",
            fontcolor=branch_border,
            color=branch_border,
            style="rounded",
            penwidth="1.8",
            margin="18"
        )

        # Architecture-specific resizing and preprocessing
        c.node(
            f"{p}_resize",
            f"Model-Specific Input\n{resolution}",
            fillcolor="#F4F7FA",
            color=branch_border
        )

        # CNN feature extraction
        c.node(
            f"{p}_cnn",
            f"{cnn_name}\nDeep Feature Extraction",
            fillcolor=branch_color,
            color=branch_border
        )

        # Squeeze-and-Excitation attention
        c.node(
            f"{p}_se",
            "SE Attention Module\nSqueeze | Excitation\nChannel Recalibration",
            fillcolor="#FFF2CC",
            color="#BF9000"
        )

        # Global average pooling and normalization
        c.node(
            f"{p}_gap",
            "Global Average Pooling\nBatch Normalization",
            fillcolor="#E4DFEC",
            color="#8064A2"
        )

        # Common-dimensional feature projection
        c.node(
            f"{p}_projection",
            "Dense Feature Projection\n512-D Representation",
            fillcolor="#DDEBF7",
            color="#4472C4"
        )

        # Internal branch connections
        c.edge(f"{p}_resize", f"{p}_cnn")
        c.edge(f"{p}_cnn", f"{p}_se")
        c.edge(f"{p}_se", f"{p}_gap")
        c.edge(f"{p}_gap", f"{p}_projection")

    # Connect shared preprocessing to each branch input
    dot.edge("preprocessing", f"{p}_resize")

    feature_nodes.append(f"{p}_projection")


# STEP 7: Align the three CNN branches horizontally
# This helps produce a balanced three-column architecture.
for stage in [
    ["x_resize", "d_resize", "m_resize"],
    ["x_cnn", "d_cnn", "m_cnn"],
    ["x_se", "d_se", "m_se"],
    ["x_gap", "d_gap", "m_gap"],
    ["x_projection", "d_projection", "m_projection"]
]:
    with dot.subgraph() as s:
        s.attr(rank="same")

        for node_id in stage:
            s.node(node_id)


# STEP 8: Feature concatenation
dot.node(
    "concatenation",
    "Multi-Branch Feature Concatenation\n"
    "1536-D Combined Feature Representation",
    fillcolor="#FCE4D6",
    color="#C65911",
    penwidth="1.8",
    fontsize="12"
)

# Connect the three projected feature representations
for node_id in feature_nodes:
    dot.edge(node_id, "concatenation")


# STEP 9: Adaptive branch-weight generation
dot.node(
    "weight_generation",
    "Adaptive Branch-Weight Generation\n"
    "Dense Layer (3) + Softmax\n"
    "Three Learnable Branch Weights",
    fillcolor="#FFF2CC",
    color="#BF9000",
    penwidth="1.8",
    fontsize="11"
)

dot.edge("concatenation", "weight_generation")


# STEP 10: Adaptive weighted feature fusion
dot.node(
    "fusion",
    "Adaptive Multi-Scale Feature Fusion\n"
    "Weighted Combination of Three Branch Features\n"
    "512-D Fused Feature Representation",
    fillcolor="#D9EAD3",
    color="#38761D",
    penwidth="2.0",
    fontsize="12"
)

# The learned weights and original branch features
# contribute to the weighted fusion operation.
dot.edge("weight_generation", "fusion")

for node_id in feature_nodes:
    dot.edge(
        node_id,
        "fusion",
        style="dashed",
        color="#78909C",
        penwidth="1.2",
        constraint="false"
    )


# STEP 11: Classification module
dot.node(
    "dense512",
    "Fully Connected Layer\n512 Units + ReLU",
    fillcolor="#EAF2F8",
    color="#2874A6"
)

dot.node(
    "dropout",
    "Dropout",
    fillcolor="#FCE4D6",
    color="#C65911"
)

dot.node(
    "dense256",
    "Fully Connected Layer\n256 Units + ReLU",
    fillcolor="#EAF2F8",
    color="#2874A6"
)

dot.node(
    "softmax",
    "Softmax Classification\nBead | Fiber | Fragment",
    fillcolor="#D9EAD3",
    color="#38761D",
    penwidth="2.0",
    fontsize="12"
)

# Connect the classification layers
dot.edge("fusion", "dense512")
dot.edge("dense512", "dropout")
dot.edge("dropout", "dense256")
dot.edge("dense256", "softmax")


# STEP 12: Grad-CAM explainability
# This is a separate interpretability output.
# It is not part of the forward classification sequence.

dot.node(
    "gradcam",
    "Grad-CAM Explainability\n"
    "Class-Discriminative Region Visualization\n"
    "Branch-Wise Feature Map Analysis",
    fillcolor="#E4DFEC",
    color="#8064A2",
    penwidth="1.8",
    fontsize="11"
)

# Grad-CAM uses the predicted class score and selected
# convolutional feature maps from the trained branches.
dot.edge(
    "softmax",
    "gradcam",
    style="dashed",
    color="#8064A2",
    label="Class score"
)

for node_id in ["x_cnn", "d_cnn", "m_cnn"]:
    dot.edge(
        node_id,
        "gradcam",
        style="dashed",
        color="#A78BC4",
        constraint="false"
    )


# STEP 13: Save the architecture in multiple formats
# PDF and SVG are vector formats suitable for manuscript figures.

png_file = dot.render(
    filename="AMAFFNet_SE_SCI_Architecture",
    format="png",
    cleanup=True
)

pdf_file = dot.render(
    filename="AMAFFNet_SE_SCI_Architecture",
    format="pdf",
    cleanup=True
)

svg_file = dot.render(
    filename="AMAFFNet_SE_SCI_Architecture",
    format="svg",
    cleanup=True
)

# STEP 14: Display the generated PNG
display(Image(filename=png_file))

# STEP 15: Report saved files
print("\nArchitecture files generated:")
print("High-resolution PNG:", png_file)
print("Vector PDF:", pdf_file)
print("Vector SVG:", svg_file)
```

```python

# ================================================================
# AMAFF-Net_SE: Final SCI Manuscript Architecture
# ================================================================

# Install Graphviz
!apt-get -qq install graphviz
!pip install -q graphviz

from graphviz import Digraph
from IPython.display import display, Image

# Create directed architecture
dot = Digraph("AMAFFNet_SE", format="png")

# Overall figure formatting
dot.attr(
    rankdir="TB",
    splines="ortho",
    nodesep="0.55",
    ranksep="0.55",
    pad="0.3",
    bgcolor="white",
    dpi="600",
    labelloc="t",
    label="Proposed Adaptive Multi-scale Attention Feature Fusion Network (AMAFF-Net_SE)",
    fontsize="21",
    fontname="Arial Bold",
    fontcolor="#24364B"
)

# Common node formatting
dot.attr(
    "node",
    shape="box",
    style="rounded,filled",
    fontname="Arial",
    fontsize="10",
    margin="0.17,0.11",
    penwidth="1.4",
    color="#53657A",
    fontcolor="#24364B"
)

# Common edge formatting
dot.attr(
    "edge",
    color="#526579",
    penwidth="1.3",
    arrowsize="0.75"
)

# ------------------------------------------------
# 1. SEM input and preprocessing
# ------------------------------------------------

dot.node(
    "input",
    "Input SEM Micrograph",
    fillcolor="#EAF2F8",
    color="#2874A6"
)

dot.node(
    "prep",
    "Model-Specific Image Preprocessing",
    fillcolor="#E8F5E9",
    color="#548235"
)

dot.edge("input", "prep")

# ------------------------------------------------
# 2. Three parallel CNN feature extraction branches
# ------------------------------------------------

branches = [
    ("x", "Xception", "299 x 299", "#DCEAF7", "#2874A6"),
    ("d", "DenseNet201", "224 x 224", "#E2F0D9", "#548235"),
    ("m", "MobileNetV2", "160 x 160", "#FCE4D6", "#C65911")
]

projection_nodes = []

for prefix, name, resolution, fill, border in branches:

    # Each branch is enclosed in a separate cluster
    with dot.subgraph(name="cluster_" + prefix) as c:

        c.attr(
            label=name + " Branch",
            style="rounded",
            color=border,
            penwidth="1.7",
            fontsize="12",
            fontname="Arial Bold",
            fontcolor=border,
            margin="15"
        )

        # Model-specific input
        c.node(
            prefix + "_input",
            "Input Resize\n" + resolution,
            fillcolor="#F4F7FA",
            color=border
        )

        # CNN backbone
        c.node(
            prefix + "_cnn",
            name + "\nFeature Extraction",
            fillcolor=fill,
            color=border
        )

        # SE attention block
        c.node(
            prefix + "_se",
            "SE Attention Module",
            fillcolor="#FFF2CC",
            color="#BF9000"
        )

        # Feature aggregation
        c.node(
            prefix + "_gap",
            "Global Average Pooling",
            fillcolor="#E4DFEC",
            color="#8064A2"
        )

        # Feature normalization
        c.node(
            prefix + "_bn",
            "Batch Normalization",
            fillcolor="#E4DFEC",
            color="#8064A2"
        )

        # Common feature projection
        c.node(
            prefix + "_proj",
            "Feature Projection",
            fillcolor="#DDEBF7",
            color="#4472C4"
        )

        # Sequential branch connections
        c.edge(prefix + "_input", prefix + "_cnn")
        c.edge(prefix + "_cnn", prefix + "_se")
        c.edge(prefix + "_se", prefix + "_gap")
        c.edge(prefix + "_gap", prefix + "_bn")
        c.edge(prefix + "_bn", prefix + "_proj")

    # Connect common preprocessing to each branch
    dot.edge("prep", prefix + "_input")

    projection_nodes.append(prefix + "_proj")


# ------------------------------------------------
# 3. Align parallel CNN branches
# ------------------------------------------------

stages = [
    ["x_input", "d_input", "m_input"],
    ["x_cnn", "d_cnn", "m_cnn"],
    ["x_se", "d_se", "m_se"],
    ["x_gap", "d_gap", "m_gap"],
    ["x_bn", "d_bn", "m_bn"],
    ["x_proj", "d_proj", "m_proj"]
]

for stage in stages:
    with dot.subgraph() as s:
        s.attr(rank="same")
        for node in stage:
            s.node(node)


# ------------------------------------------------
# 4. Feature concatenation
# ------------------------------------------------

dot.node(
    "concat",
    "Feature Concatenation",
    fillcolor="#FCE4D6",
    color="#C65911",
    penwidth="1.7",
    fontsize="11"
)

for node in projection_nodes:
    dot.edge(node, "concat")


# ------------------------------------------------
# 5. Adaptive branch-weight generation
# ------------------------------------------------

dot.node(
    "weights",
    "Adaptive Branch-Weight Generation\nDense Layer + Softmax",
    fillcolor="#FFF2CC",
    color="#BF9000",
    penwidth="1.7"
)

dot.edge("concat", "weights")


# ------------------------------------------------
# 6. Adaptive weighted feature fusion
# ------------------------------------------------

dot.node(
    "fusion",
    "Adaptive Weighted Feature Fusion",
    fillcolor="#D9EAD3",
    color="#38761D",
    penwidth="2.0",
    fontsize="12"
)

dot.edge("weights", "fusion")

# Dashed connections indicate the original branch
# features are also supplied to the fusion operation.
for node in projection_nodes:
    dot.edge(
        node,
        "fusion",
        style="dashed",
        color="#78909C",
        constraint="false"
    )


# ------------------------------------------------
# 7. Final classification
# ------------------------------------------------

dot.node(
    "classifier",
    "Final Classification\nBead | Fiber | Fragment",
    fillcolor="#D9EAD3",
    color="#38761D",
    penwidth="2.0",
    fontsize="12"
)

dot.edge("fusion", "classifier")


# ------------------------------------------------
# 8. Grad-CAM explainability
# ------------------------------------------------

dot.node(
    "gradcam",
    "Grad-CAM Explainability\n"
    "Class-Discriminative Region Visualization",
    fillcolor="#E4DFEC",
    color="#8064A2",
    penwidth="1.7"
)

# Separate dashed explainability connections
dot.edge(
    "classifier",
    "gradcam",
    style="dashed",
    color="#8064A2",
    label="Class score"
)

for node in ["x_cnn", "d_cnn", "m_cnn"]:
    dot.edge(
        node,
        "gradcam",
        style="dashed",
        color="#A78BC4",
        constraint="false"
    )


# ------------------------------------------------
# 9. Export manuscript-ready files
# ------------------------------------------------

png_file = dot.render(
    "AMAFFNet_SE_Final_Architecture",
    format="png",
    cleanup=True
)

pdf_file = dot.render(
    "AMAFFNet_SE_Final_Architecture",
    format="pdf",
    cleanup=True
)

svg_file = dot.render(
    "AMAFFNet_SE_Final_Architecture",
    format="svg",
    cleanup=True
)

# Display the final architecture
display(Image(filename=png_file))

print("Generated successfully:")
print("PNG:", png_file)
print("PDF:", pdf_file)
print("SVG:", svg_file)
```

```python

# ================================================================
# AMAFF-Net_SE: Clean SCI-Style Architecture Diagram
# Landscape layout with aligned parallel CNN branches
# ================================================================

import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch, FancyArrowPatch

# ------------------------------------------------
# 1. Create a landscape canvas
# ------------------------------------------------

fig, ax = plt.subplots(figsize=(19, 10))

ax.set_xlim(0, 19)
ax.set_ylim(0, 10)
ax.axis("off")

# ------------------------------------------------
# 2. Define a consistent visual style
# ------------------------------------------------

COLORS = {
    "input": "#EAF1F8",
    "branch": "#F3F6FA",
    "attention": "#FFF2D8",
    "processing": "#EAF3EE",
    "fusion": "#E7EFF8",
    "classifier": "#E5F0E7",
    "explain": "#F0EAF6",
    "border": "#526579",
    "text": "#24364B",
    "arrow": "#526579"
}

def add_box(x, y, width, height, text,
            facecolor, fontsize=9.5,
            edgecolor=None, linewidth=1.3,
            weight="normal"):

    """Draw a rounded rectangular architecture block."""

    box = FancyBboxPatch(
        (x, y),
        width,
        height,
        boxstyle="round,pad=0.04,rounding_size=0.12",
        facecolor=facecolor,
        edgecolor=edgecolor or COLORS["border"],
        linewidth=linewidth
    )

    ax.add_patch(box)

    ax.text(
        x + width / 2,
        y + height / 2,
        text,
        ha="center",
        va="center",
        fontsize=fontsize,
        color=COLORS["text"],
        fontweight=weight,
        linespacing=1.35
    )

    return (x, y, width, height)


def add_arrow(x1, y1, x2, y2,
              style="-",
              color=None,
              linewidth=1.3,
              connection="arc3"):

    """Draw a directed connector between architecture blocks."""

    arrow = FancyArrowPatch(
        (x1, y1),
        (x2, y2),
        arrowstyle="-|>",
        mutation_scale=12,
        linewidth=linewidth,
        linestyle=style,
        color=color or COLORS["arrow"],
        connectionstyle=connection
    )

    ax.add_patch(arrow)


# ------------------------------------------------
# 3. Title
# ------------------------------------------------

ax.text(
    9.5, 9.65,
    "Proposed Adaptive Multi-scale Attention Feature Fusion Network",
    ha="center",
    va="center",
    fontsize=17,
    fontweight="bold",
    color=COLORS["text"]
)

ax.text(
    9.5, 9.28,
    "(AMAFF-Net_SE)",
    ha="center",
    fontsize=13,
    color=COLORS["text"]
)


# ------------------------------------------------
# 4. Input and common preprocessing
# ------------------------------------------------

add_box(
    7.2, 8.35, 4.6, 0.55,
    "Input SEM Micrograph",
    COLORS["input"], 11, weight="bold"
)

add_arrow(9.5, 8.35, 9.5, 7.98)

add_box(
    6.6, 7.38, 5.8, 0.58,
    "Model-Specific Image Preprocessing",
    COLORS["processing"], 10, weight="bold"
)

# ------------------------------------------------
# 5. Three aligned parallel feature extraction branches
# ------------------------------------------------

# Each row represents one independent CNN branch.
# The same original SEM image is preprocessed for each model.

branch_rows = [
    {
        "name": "Xception",
        "resolution": "299 x 299",
        "y": 6.25
    },
    {
        "name": "DenseNet201",
        "resolution": "224 x 224",
        "y": 4.95
    },
    {
        "name": "MobileNetV2",
        "resolution": "160 x 160",
        "y": 3.65
    }
]

# Block positions and dimensions are identical across branches.
positions = {
    "input": (0.45, 1.75),
    "cnn": (2.65, 2.00),
    "se": (5.05, 1.65),
    "gap": (7.15, 1.75),
    "bn": (9.35, 1.65),
    "projection": (11.45, 2.00)
}

branch_output_centers = []

for branch in branch_rows:

    y = branch["y"]
    name = branch["name"]
    resolution = branch["resolution"]

    # Branch label
    ax.text(
        0.35,
        y + 0.47,
        name,
        ha="left",
        va="bottom",
        fontsize=10,
        fontweight="bold",
        color=COLORS["text"]
    )

    # Model-specific input
    x, w = positions["input"]
    add_box(
        x, y, w, 0.55,
        f"Input\n{resolution}",
        COLORS["input"]
    )

    # CNN feature extraction
    x, w = positions["cnn"]
    add_box(
        x, y, w, 0.55,
        f"{name}\nFeature Extraction",
        COLORS["branch"],
        weight="bold"
    )

    # SE attention
    x, w = positions["se"]
    add_box(
        x, y, w, 0.55,
        "SE Attention",
        COLORS["attention"],
        weight="bold"
    )

    # Global Average Pooling
    x, w = positions["gap"]
    add_box(
        x, y, w, 0.55,
        "Global Average\nPooling",
        COLORS["processing"]
    )

    # Batch Normalization
    x, w = positions["bn"]
    add_box(
        x, y, w, 0.55,
        "Batch\nNormalization",
        COLORS["processing"]
    )

    # Feature projection
    x, w = positions["projection"]
    add_box(
        x, y, w, 0.55,
        "Feature Projection",
        COLORS["fusion"]
    )

    # Connect the blocks in a straight horizontal sequence
    centers_y = y + 0.275

    add_arrow(2.20, centers_y, 2.65, centers_y)
    add_arrow(4.65, centers_y, 5.05, centers_y)
    add_arrow(6.70, centers_y, 7.15, centers_y)
    add_arrow(8.90, centers_y, 9.35, centers_y)
    add_arrow(11.00, centers_y, 11.45, centers_y)

    # Store branch output location
    branch_output_centers.append(
        (13.45, centers_y)
    )

# Connect common preprocessing to the three branch inputs
# The shared vertical line avoids crossing arrows.

ax.plot(
    [9.5, 9.5],
    [7.38, 6.525],
    color=COLORS["arrow"],
    linewidth=1.3
)

ax.plot(
    [1.325, 9.5],
    [6.525, 6.525],
    color=COLORS["arrow"],
    linewidth=1.3
)

ax.plot(
    [1.325, 1.325],
    [4.025, 6.525],
    color=COLORS["arrow"],
    linewidth=1.3
)

# Small branch connection markers
for _, cy in branch_output_centers:
    ax.scatter(
        [1.325], [cy],
        s=12,
        color=COLORS["arrow"],
        zorder=5
    )

# ------------------------------------------------
# 6. Feature concatenation
# ------------------------------------------------

add_box(
    14.05, 5.05, 2.3, 0.65,
    "Feature\nConcatenation",
    COLORS["fusion"],
    10,
    weight="bold"
)

# Branch outputs connect to a clean collection point.
# The collection point routes the branch features to concatenation.

junction_x = 13.78

for bx, by in branch_output_centers:

    add_arrow(
        bx, by,
        junction_x, by
    )

ax.plot(
    [junction_x, junction_x],
    [branch_output_centers[2][1],
     branch_output_centers[0][1]],
    color=COLORS["arrow"],
    linewidth=1.3
)

add_arrow(
    junction_x,
    5.25,
    14.05,
    5.25
)

# ------------------------------------------------
# 7. Adaptive branch-weight generation
# ------------------------------------------------

add_box(
    14.05, 3.95, 2.3, 0.65,
    "Adaptive Branch-Weight\nGeneration",
    COLORS["attention"],
    9.5,
    weight="bold"
)

add_arrow(15.20, 5.05, 15.20, 4.60)

# ------------------------------------------------
# 8. Adaptive weighted feature fusion
# ------------------------------------------------

add_box(
    14.05, 2.85, 2.3, 0.65,
    "Adaptive Weighted\nFeature Fusion",
    COLORS["processing"],
    9.5,
    weight="bold"
)

add_arrow(15.20, 3.95, 15.20, 3.50)

# ------------------------------------------------
# 9. Final classification
# ------------------------------------------------

add_box(
    14.05, 1.75, 2.3, 0.65,
    "Final Classification\nBead | Fiber | Fragment",
    COLORS["classifier"],
    9.5,
    weight="bold"
)

add_arrow(15.20, 2.85, 15.20, 2.40)

# ------------------------------------------------
# 10. Grad-CAM: Separate explainability panel
# ------------------------------------------------

add_box(
    6.1, 0.45, 6.8, 0.70,
    "Grad-CAM Explainability\n"
    "Class-Discriminative Region Visualization",
    COLORS["explain"],
    10,
    edgecolor="#8064A2",
    linewidth=1.6,
    weight="bold"
)

# Separate dashed connection from classification output.
add_arrow(
    14.05, 2.00,
    12.90, 0.80,
    style="--",
    color="#8064A2",
    linewidth=1.2,
    connection="arc3,rad=-0.18"
)

# ------------------------------------------------
# 11. Final formatting and export
# ------------------------------------------------

plt.tight_layout()

# Save high-resolution image and vector formats
plt.savefig(
    "AMAFFNet_SE_Clean_Architecture.png",
    dpi=600,
    bbox_inches="tight",
    facecolor="white"
)

plt.savefig(
    "AMAFFNet_SE_Clean_Architecture.pdf",
    bbox_inches="tight",
    facecolor="white"
)

plt.savefig(
    "AMAFFNet_SE_Clean_Architecture.svg",
    bbox_inches="tight",
    facecolor="white"
)

plt.show()

print("Architecture exported successfully.")
```

```python

# ================================================================
# CELL 7: COMPLETE EIGHT-MODEL RESULTS AND DISCUSSION FIGURES
# Uses saved five-fold predictions and metrics only.
# No model training or prediction regeneration.
# Output: 600-DPI PNG + PDF + CSV tables
# ================================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from pathlib import Path
from sklearn.metrics import (
    confusion_matrix,
    classification_report,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    cohen_kappa_score,
    matthews_corrcoef,
    roc_auc_score,
    roc_curve,
    auc
)
from sklearn.preprocessing import label_binarize

# ------------------------------------------------
# 1. DIRECTORY SETTINGS
# ------------------------------------------------

BASE_DIR = Path("/content/drive/MyDrive/Microplastics_SEM")

BACKBONE_DIR = (
    BASE_DIR / "AMAFFNet_Results" /
    "Backbone_SE_NoSE_Evaluation"
)

FUSION_DIR = BASE_DIR / "AMAFFNet_Fusion_Results"

OUTPUT_DIR = (
    BASE_DIR / "AMAFFNet_Visualization"
)

CM_DIR = OUTPUT_DIR / "Confusion_Matrices"
ROC_DIR = OUTPUT_DIR / "ROC_Curves"
GRAPH_DIR = OUTPUT_DIR / "Comparison_Graphs"
CLASS_DIR = OUTPUT_DIR / "Classwise_Performance"
TABLE_DIR = OUTPUT_DIR / "Results_Tables"

for folder in [
    CM_DIR, ROC_DIR, GRAPH_DIR, CLASS_DIR, TABLE_DIR
]:
    folder.mkdir(parents=True, exist_ok=True)

CLASS_NAMES = ["Bead", "Fiber", "Fragment"]
CLASS_TO_ID = {
    "Bead": 0,
    "Fiber": 1,
    "Fragment": 2
}

MODEL_NAMES = [
    "Xception_NoSE",
    "Xception_SE",
    "DenseNet201_NoSE",
    "DenseNet201_SE",
    "MobileNetV2_NoSE",
    "MobileNetV2_SE",
    "AMAFFNet_NoSE",
    "AMAFFNet_SE"
]

# ------------------------------------------------
# 2. SAVED PREDICTION FILES
# ------------------------------------------------

PREDICTION_FILES = {
    "Xception_NoSE": [
        BASE_DIR / "AMAFFNet_Results" /
        "Xception" / "Xception_All_Fold_Predictions.csv"
    ],

    "Xception_SE": [
        BASE_DIR / "AMAFFNet_Results" /
        "Xception_SE" / "Xception_SE_All_Fold_Predictions.csv"
    ],

    "DenseNet201_NoSE": [
        BASE_DIR / "AMAFFNet_Results" /
        "DenseNet201_NoSE" /
        "DenseNet201_NoSE_All_Fold_Predictions.csv"
    ],

    "DenseNet201_SE": [
        BASE_DIR / "AMAFFNet_Results" /
        "DenseNet201" / "DenseNet201_All_Fold_Predictions.csv"
    ],

    "MobileNetV2_NoSE": [
        BASE_DIR / "AMAFFNet_Results" /
        "MobileNetV2_NoSE" /
        "MobileNetV2_NoSE_All_Fold_Predictions.csv"
    ],

    "MobileNetV2_SE": [
        BASE_DIR / "AMAFFNet_Results" /
        "MobileNetV2" /
        "MobileNetV2_All_Fold_Predictions.csv"
    ],

    # AMAFF-Net predictions are collected from the five fold folders.
    "AMAFFNet_NoSE": [
        FUSION_DIR / "AMAFFNet_NoSE" /
        f"Fold_{fold}" / "Test_Predictions.csv"
        for fold in range(1, 6)
    ],

    "AMAFFNet_SE": [
        FUSION_DIR / "AMAFFNet_SE" /
        f"Fold_{fold}" / "Test_Predictions.csv"
        for fold in range(1, 6)
    ]
}

# ------------------------------------------------
# 3. STANDARDIZE CLASS LABELS
# ------------------------------------------------

def standardize_labels(series):

    standardized = []

    for value in series:

        # Convert numeric labels to integer class IDs.
        if isinstance(value, (int, np.integer)):
            standardized.append(int(value))
            continue

        if isinstance(value, (float, np.floating)):
            standardized.append(int(value))
            continue

        value = str(value).strip()

        if value in CLASS_TO_ID:
            standardized.append(CLASS_TO_ID[value])

        elif value in ["0", "1", "2"]:
            standardized.append(int(value))

        else:
            raise ValueError(
                f"Unrecognized class label: {value}"
            )

    return np.asarray(standardized, dtype=int)


def find_column(dataframe, candidates):

    for column in candidates:
        if column in dataframe.columns:
            return column

    raise KeyError(
        f"None of these columns were found: {candidates}\n"
        f"Available columns: {list(dataframe.columns)}"
    )


# ------------------------------------------------
# 4. LOAD AND STANDARDIZE PREDICTIONS
# ------------------------------------------------

all_predictions = {}

for model_name, file_list in PREDICTION_FILES.items():

    print("\nLoading:", model_name)

    fold_frames = []

    for file_path in file_list:

        if not file_path.exists():
            raise FileNotFoundError(
                f"Prediction file missing:\n{file_path}"
            )

        fold_frames.append(
            pd.read_csv(file_path)
        )

    raw_df = pd.concat(
        fold_frames,
        ignore_index=True
    )

    # Identify actual and predicted label columns.
    true_column = find_column(
        raw_df,
        [
            "True_Class",
            "Actual_Class",
            "Class",
            "True_Label",
            "Actual_Label"
        ]
    )

    predicted_column = find_column(
        raw_df,
        [
            "Predicted_Class",
            "Predicted_Label"
        ]
    )

    # Identify probability columns.
    probability_columns = [
        [
            "Probability_Bead",
            "Bead_Probability"
        ],
        [
            "Probability_Fiber",
            "Fiber_Probability"
        ],
        [
            "Probability_Fragment",
            "Fragment_Probability"
        ]
    ]

    selected_probability_columns = []

    for candidates in probability_columns:
        selected_probability_columns.append(
            find_column(raw_df, candidates)
        )

    # Standardize all model outputs into one format.
    standardized_df = pd.DataFrame({
        "True_Label": standardize_labels(
            raw_df[true_column]
        ),
        "Predicted_Label": standardize_labels(
            raw_df[predicted_column]
        )
    })

    standardized_df["Probability_Bead"] = (
        raw_df[selected_probability_columns[0]].astype(float)
    )

    standardized_df["Probability_Fiber"] = (
        raw_df[selected_probability_columns[1]].astype(float)
    )

    standardized_df["Probability_Fragment"] = (
        raw_df[selected_probability_columns[2]].astype(float)
    )

    # Confirm the expected number of out-of-fold predictions.
    if len(standardized_df) != 237:
        raise ValueError(
            f"{model_name}: Expected 237 predictions, "
            f"but found {len(standardized_df)}."
        )

    if standardized_df.isnull().any().any():
        raise ValueError(
            f"Missing values found in {model_name} predictions."
        )

    probabilities = standardized_df[
        [
            "Probability_Bead",
            "Probability_Fiber",
            "Probability_Fragment"
        ]
    ].to_numpy()

    if not np.allclose(
        probabilities.sum(axis=1), 1.0, atol=1e-4
    ):
        raise ValueError(
            f"Probability sums are invalid for {model_name}."
        )

    all_predictions[model_name] = standardized_df

    print("Predictions:", len(standardized_df))
    print("Prediction file verification: PASSED")


# ------------------------------------------------
# 5. GENERATE EIGHT CONFUSION MATRICES
# ------------------------------------------------

pooled_metric_rows = []
classwise_rows = []

for model_name in MODEL_NAMES:

    df = all_predictions[model_name]

    y_true = df["True_Label"].to_numpy()
    y_pred = df["Predicted_Label"].to_numpy()

    probabilities = df[
        [
            "Probability_Bead",
            "Probability_Fiber",
            "Probability_Fragment"
        ]
    ].to_numpy()

    cm = confusion_matrix(
        y_true,
        y_pred,
        labels=[0, 1, 2]
    )

    # Save the numerical confusion matrix.
    pd.DataFrame(
        cm,
        index=CLASS_NAMES,
        columns=CLASS_NAMES
    ).to_csv(
        CM_DIR / f"{model_name}_Confusion_Matrix.csv"
    )

    # Draw a publication-quality confusion matrix.
    fig, ax = plt.subplots(figsize=(7, 6))

    image_plot = ax.imshow(
        cm,
        interpolation="nearest",
        cmap="Blues"
    )

    fig.colorbar(image_plot, ax=ax, fraction=0.046)

    ax.set(
        xticks=np.arange(3),
        yticks=np.arange(3),
        xticklabels=CLASS_NAMES,
        yticklabels=CLASS_NAMES,
        xlabel="Predicted Class",
        ylabel="Actual Class",
        title=f"Confusion Matrix: {model_name}"
    )

    threshold = cm.max() / 2.0

    for i in range(3):
        for j in range(3):

            ax.text(
                j, i, str(cm[i, j]),
                ha="center",
                va="center",
                fontsize=13,
                color=(
                    "white"
                    if cm[i, j] > threshold
                    else "black"
                )
            )

    plt.tight_layout()

    fig.savefig(
        CM_DIR / f"{model_name}_Confusion_Matrix.png",
        dpi=600,
        bbox_inches="tight"
    )

    fig.savefig(
        CM_DIR / f"{model_name}_Confusion_Matrix.pdf",
        bbox_inches="tight"
    )

    plt.close(fig)

    # ------------------------------------------------
    # CLASSIFICATION REPORT
    # ------------------------------------------------

    report = classification_report(
        y_true,
        y_pred,
        labels=[0, 1, 2],
        target_names=CLASS_NAMES,
        output_dict=True,
        zero_division=0
    )

    pd.DataFrame(report).transpose().to_csv(
        CLASS_DIR / f"{model_name}_Classification_Report.csv"
    )

    for class_name in CLASS_NAMES:

        classwise_rows.append({
            "Model": model_name,
            "Class": class_name,
            "Precision": report[class_name]["precision"],
            "Recall": report[class_name]["recall"],
            "F1_Score": report[class_name]["f1-score"],
            "Support": report[class_name]["support"]
        })

    # ------------------------------------------------
    # POOLED OUT-OF-FOLD METRICS
    # ------------------------------------------------

    pooled_metric_rows.append({
        "Model": model_name,
        "Accuracy": accuracy_score(y_true, y_pred),
        "Precision_Macro": precision_score(
            y_true, y_pred, average="macro",
            zero_division=0
        ),
        "Recall_Macro": recall_score(
            y_true, y_pred, average="macro",
            zero_division=0
        ),
        "F1_Macro": f1_score(
            y_true, y_pred, average="macro",
            zero_division=0
        ),
        "Balanced_Accuracy": balanced_accuracy_score(
            y_true, y_pred
        ),
        "Cohens_Kappa": cohen_kappa_score(
            y_true, y_pred
        ),
        "MCC": matthews_corrcoef(y_true, y_pred),
        "ROC_AUC_Macro_OVR": roc_auc_score(
            y_true,
            probabilities,
            multi_class="ovr",
            average="macro"
        )
    })

print("\nAll eight confusion matrices generated.")


# ------------------------------------------------
# 6. GENERATE INDIVIDUAL MULTICLASS ROC CURVES
# ------------------------------------------------

y_color_labels = label_binarize(
    np.arange(3),
    classes=[0, 1, 2]
)

fpr_grid = np.linspace(0, 1, 1000)

combined_roc_rows = []

for model_name in MODEL_NAMES:

    df = all_predictions[model_name]

    y_true = df["True_Label"].to_numpy()

    probabilities = df[
        [
            "Probability_Bead",
            "Probability_Fiber",
            "Probability_Fragment"
        ]
    ].to_numpy()

    y_binary = label_binarize(
        y_true,
        classes=[0, 1, 2]
    )

    fig, ax = plt.subplots(figsize=(8, 7))

    interpolated_tprs = []

    for class_id, class_name in enumerate(CLASS_NAMES):

        fpr, tpr, _ = roc_curve(
            y_binary[:, class_id],
            probabilities[:, class_id]
        )

        class_auc = auc(fpr, tpr)

        ax.plot(
            fpr,
            tpr,
            linewidth=2,
            label=f"{class_name} (AUC = {class_auc:.4f})"
        )

        interpolated_tpr = np.interp(
            fpr_grid,
            fpr,
            tpr
        )

        interpolated_tpr[0] = 0.0

        interpolated_tprs.append(
            interpolated_tpr
        )

    # Calculate the macro-average ROC curve.
    mean_tpr = np.mean(
        interpolated_tprs,
        axis=0
    )

    mean_tpr[-1] = 1.0

    macro_auc = auc(
        fpr_grid,
        mean_tpr
    )

    ax.plot(
        fpr_grid,
        mean_tpr,
        linestyle="--",
        linewidth=2.5,
        label=f"Macro-average (AUC = {macro_auc:.4f})"
    )

    ax.plot(
        [0, 1], [0, 1],
        linestyle=":",
        linewidth=1.5,
        label="Random Classifier"
    )

    ax.set(
        xlabel="False Positive Rate",
        ylabel="True Positive Rate",
        title=f"Multiclass ROC Curve: {model_name}",
        xlim=(0, 1),
        ylim=(0, 1.02)
    )

    ax.legend(loc="lower right", fontsize=9)
    ax.grid(alpha=0.2)

    plt.tight_layout()

    fig.savefig(
        ROC_DIR / f"{model_name}_ROC.png",
        dpi=600,
        bbox_inches="tight"
    )

    fig.savefig(
        ROC_DIR / f"{model_name}_ROC.pdf",
        bbox_inches="tight"
    )

    plt.close(fig)

    combined_roc_rows.append({
        "Model": model_name,
        "FPR": fpr_grid,
        "TPR": mean_tpr,
        "AUC": macro_auc
    })

print("Eight individual ROC curves generated.")


# ------------------------------------------------
# 7. COMBINED ROC COMPARISON OF EIGHT MODELS
# ------------------------------------------------

fig, ax = plt.subplots(figsize=(10, 8))

for result in combined_roc_rows:

    ax.plot(
        result["FPR"],
        result["TPR"],
        linewidth=2,
        label=(
            f'{result["Model"]} '
            f'(AUC = {result["AUC"]:.4f})'
        )
    )

ax.plot(
    [0, 1], [0, 1],
    linestyle=":",
    linewidth=1.5,
    label="Random Classifier"
)

ax.set(
    xlabel="False Positive Rate",
    ylabel="True Positive Rate",
    title="Macro-Averaged ROC Comparison of Eight Models",
    xlim=(0, 1),
    ylim=(0, 1.02)
)

ax.legend(
    loc="lower right",
    fontsize=8
)

ax.grid(alpha=0.2)
plt.tight_layout()

fig.savefig(
    ROC_DIR / "Eight_Model_Combined_ROC.png",
    dpi=600,
    bbox_inches="tight"
)

fig.savefig(
    ROC_DIR / "Eight_Model_Combined_ROC.pdf",
    bbox_inches="tight"
)

plt.close(fig)


# ------------------------------------------------
# 8. SAVE POOLED METRICS AND CLASS-WISE TABLES
# ------------------------------------------------

pooled_df = pd.DataFrame(pooled_metric_rows)

pooled_df.to_csv(
    TABLE_DIR / "Eight_Model_Pooled_OOF_Metrics.csv",
    index=False
)

classwise_df = pd.DataFrame(classwise_rows)

classwise_df.to_csv(
    CLASS_DIR / "Eight_Model_Classwise_Performance.csv",
    index=False
)


# ------------------------------------------------
# 9. LOAD FIVE-FOLD METRICS FOR MEAN AND SD
# ------------------------------------------------

backbone_metrics_file = (
    BACKBONE_DIR / "All_Six_Models_5Fold_Metrics.csv"
)

if not backbone_metrics_file.exists():
    raise FileNotFoundError(
        f"Backbone metrics file missing:\n"
        f"{backbone_metrics_file}"
    )

backbone_metrics = pd.read_csv(
    backbone_metrics_file
)

fusion_metric_frames = []

for experiment in ["AMAFFNet_NoSE", "AMAFFNet_SE"]:

    for fold in range(1, 6):

        metric_file = (
            FUSION_DIR / experiment /
            f"Fold_{fold}" / "Fold_Metrics.csv"
        )

        if not metric_file.exists():
            raise FileNotFoundError(
                f"Missing fold metric file:\n{metric_file}"
            )

        fusion_metric_frames.append(
            pd.read_csv(metric_file)
        )

fusion_metrics = pd.concat(
    fusion_metric_frames,
    ignore_index=True
)

# Standardize model names and metric columns.
backbone_metrics = backbone_metrics.rename(
    columns={"Model": "Experiment"}
)

all_fold_metrics = pd.concat(
    [backbone_metrics, fusion_metrics],
    ignore_index=True
)

# Verify that every model has five fold records.
fold_counts = all_fold_metrics.groupby(
    "Experiment"
).size()

print("\nFive-fold metric record counts:")
print(fold_counts)

for model_name in MODEL_NAMES:
    if fold_counts.get(model_name, 0) != 5:
        raise ValueError(
            f"{model_name} does not have exactly five fold records."
        )

METRICS = [
    "Accuracy",
    "Precision_Macro",
    "Recall_Macro",
    "F1_Macro",
    "Balanced_Accuracy",
    "Cohens_Kappa",
    "MCC",
    "ROC_AUC_Macro_OVR"
]

# Calculate mean and sample standard deviation across folds.
summary_rows = []

for model_name in MODEL_NAMES:

    subset = all_fold_metrics[
        all_fold_metrics["Experiment"] == model_name
    ]

    row = {
        "Model": model_name,
        "Number_of_Folds": len(subset)
    }

    for metric in METRICS:

        row[metric + "_Mean"] = subset[metric].mean()
        row[metric + "_SD"] = subset[metric].std(ddof=1)

    summary_rows.append(row)

summary_df = pd.DataFrame(summary_rows)

summary_df.to_csv(
    TABLE_DIR / "Eight_Model_Five_Fold_Mean_SD.csv",
    index=False
)


# ------------------------------------------------
# 10. GENERATE COMPARATIVE PERFORMANCE GRAPHS
# ------------------------------------------------

metric_titles = {
    "Accuracy": "Accuracy",
    "Precision_Macro": "Macro Precision",
    "Recall_Macro": "Macro Recall",
    "F1_Macro": "Macro F1-score",
    "Balanced_Accuracy": "Balanced Accuracy",
    "Cohens_Kappa": "Cohen's Kappa",
    "MCC": "Matthews Correlation Coefficient",
    "ROC_AUC_Macro_OVR": "Macro ROC-AUC"
}

for metric in METRICS:

    means = summary_df[metric + "_Mean"].to_numpy()
    stds = summary_df[metric + "_SD"].to_numpy()

    fig, ax = plt.subplots(figsize=(12, 6))

    x_positions = np.arange(len(MODEL_NAMES))

    ax.bar(
        x_positions,
        means,
        yerr=stds,
        capsize=5,
        edgecolor="black",
        linewidth=0.7
    )

    ax.set_xticks(x_positions)
    ax.set_xticklabels(
        MODEL_NAMES,
        rotation=35,
        ha="right"
    )

    ax.set_ylabel(metric_titles[metric])
    ax.set_title(
        f"Eight-Model Comparison: {metric_titles[metric]}"
    )

    ax.set_ylim(
        max(0, min(0.0, np.nanmin(means - stds) - 0.05)),
        min(1.10, np.nanmax(means + stds) + 0.08)
    )

    ax.grid(
        axis="y",
        alpha=0.2
    )

    plt.tight_layout()

    fig.savefig(
        GRAPH_DIR / f"Eight_Model_{metric}.png",
        dpi=600,
        bbox_inches="tight"
    )

    fig.savefig(
        GRAPH_DIR / f"Eight_Model_{metric}.pdf",
        bbox_inches="tight"
    )

    plt.close(fig)


# ------------------------------------------------
# 11. GENERATE COMBINED PERFORMANCE COMPARISON
# ------------------------------------------------

selected_metrics = [
    "Accuracy",
    "Precision_Macro",
    "Recall_Macro",
    "F1_Macro",
    "Balanced_Accuracy"
]

fig, ax = plt.subplots(figsize=(13, 7))

x_positions = np.arange(len(MODEL_NAMES))
bar_width = 0.15

for index, metric in enumerate(selected_metrics):

    means = summary_df[metric + "_Mean"].to_numpy()

    ax.bar(
        x_positions + index * bar_width,
        means,
        bar_width,
        label=metric
    )

ax.set_xticks(
    x_positions + bar_width * 2
)

ax.set_xticklabels(
    MODEL_NAMES,
    rotation=35,
    ha="right"
)

ax.set_ylabel("Performance")
ax.set_title(
    "Comparative Performance of Eight Classification Models"
)

ax.legend(
    loc="lower right",
    fontsize=8
)

ax.grid(
    axis="y",
    alpha=0.2
)

plt.tight_layout()

fig.savefig(
    GRAPH_DIR / "Eight_Model_Overall_Performance.png",
    dpi=600,
    bbox_inches="tight"
)

fig.savefig(
    GRAPH_DIR / "Eight_Model_Overall_Performance.pdf",
    bbox_inches="tight"
)

plt.close(fig)


# ------------------------------------------------
# 12. SAVE COMPLETE FOLD-LEVEL METRICS
# ------------------------------------------------

all_fold_metrics.to_csv(
    TABLE_DIR / "Eight_Model_All_Five_Fold_Metrics.csv",
    index=False
)

# ------------------------------------------------
# 13. DISPLAY FINAL TABLES AND OUTPUT LOCATION
# ------------------------------------------------

print("\n" + "=" * 75)
print("EIGHT-MODEL VISUALIZATION COMPLETED")
print("=" * 75)

print("\nFIVE-FOLD MEAN ± SD RESULTS")
print(summary_df.to_string(index=False))

print("\nPOOLED OUT-OF-FOLD METRICS")
print(pooled_df.to_string(index=False))

print("\nCLASS-WISE PERFORMANCE")
print(classwise_df.to_string(index=False))

print("\nAll figures and tables saved in:")
print(OUTPUT_DIR)

print("\nPNG figures: 600 DPI")
print("PDF figures: vector format where supported")
print("CSV tables: saved")
```

```python

# ================================================================
# FIGURE 2: CONFUSION MATRICES OF SIX INDIVIDUAL BACKBONE MODELS
# Five-fold pooled out-of-fold predictions | 237 SEM images
# ================================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix

# ------------------------------------------------
# 1. Define input and output directories
# ------------------------------------------------

BASE_DIR = "/content/drive/MyDrive/Microplastics_SEM"

PREDICTION_DIR = os.path.join(BASE_DIR, "AMAFFNet_Results")

OUTPUT_DIR = os.path.join(
    BASE_DIR,
    "AMAFFNet_Visualization",
    "Confusion_Matrices"
)

os.makedirs(OUTPUT_DIR, exist_ok=True)

# ------------------------------------------------
# 2. Define the six individual model configurations
# ------------------------------------------------
# DenseNet201 is the SE configuration.
# MobileNetV2 is the SE configuration.
# Xception is the NoSE configuration.

model_files = {
    "Xception_NoSE": (
        "Xception/Xception_All_Fold_Predictions.csv"
    ),

    "Xception_SE": (
        "Xception_SE/Xception_SE_All_Fold_Predictions.csv"
    ),

    "DenseNet201_NoSE": (
        "DenseNet201_NoSE/DenseNet201_NoSE_All_Fold_Predictions.csv"
    ),

    "DenseNet201_SE": (
        "DenseNet201/DenseNet201_All_Fold_Predictions.csv"
    ),

    "MobileNetV2_NoSE": (
        "MobileNetV2_NoSE/MobileNetV2_NoSE_All_Fold_Predictions.csv"
    ),

    "MobileNetV2_SE": (
        "MobileNetV2/MobileNetV2_All_Fold_Predictions.csv"
    )
}

# Fixed class order for every confusion matrix
class_names = ["Bead", "Fiber", "Fragment"]

# Support both text labels and numeric encoded labels
label_mapping = {
    "0": "Bead",
    "1": "Fiber",
    "2": "Fragment",
    "0.0": "Bead",
    "1.0": "Fiber",
    "2.0": "Fragment",
    "bead": "Bead",
    "fiber": "Fiber",
    "fragment": "Fragment"
}

def standardize_label(value):
    """Convert labels into the fixed three-class naming scheme."""

    key = str(value).strip()

    if key.lower() in label_mapping:
        return label_mapping[key.lower()]

    if key in label_mapping:
        return label_mapping[key]

    raise ValueError(f"Unrecognized class label: {value}")


# ------------------------------------------------
# 3. Load pooled predictions and calculate matrices
# ------------------------------------------------

all_matrices = {}

for model_name, relative_path in model_files.items():

    file_path = os.path.join(PREDICTION_DIR, relative_path)

    print("\nProcessing:", model_name)

    if not os.path.exists(file_path):
        raise FileNotFoundError(
            f"Prediction file not found:\n{file_path}"
        )

    df = pd.read_csv(file_path)

    # Automatically identify the actual-label column
    true_candidates = [
        "True_Class",
        "Actual_Class",
        "Class",
        "True_Label",
        "Actual_Label"
    ]

    true_col = next(
        (col for col in true_candidates if col in df.columns),
        None
    )

    # Identify the predicted-label column
    pred_candidates = [
        "Predicted_Class",
        "Predicted_Label"
    ]

    pred_col = next(
        (col for col in pred_candidates if col in df.columns),
        None
    )

    if true_col is None or pred_col is None:
        raise ValueError(
            f"Required label columns not found in {file_path}\n"
            f"Available columns: {df.columns.tolist()}"
        )

    # Standardize the labels
    y_true = df[true_col].apply(standardize_label)
    y_pred = df[pred_col].apply(standardize_label)

    # Check prediction completeness
    if len(df) != 237:
        raise ValueError(
            f"{model_name}: Expected 237 predictions, "
            f"but found {len(df)}."
        )

    if y_true.isna().any() or y_pred.isna().any():
        raise ValueError(
            f"Missing labels detected in {model_name}"
        )

    # Calculate pooled out-of-fold confusion matrix
    cm = confusion_matrix(
        y_true,
        y_pred,
        labels=class_names
    )

    all_matrices[model_name] = cm

    print("Total predictions:", len(df))
    print("Confusion matrix:\n", cm)

# ------------------------------------------------
# 4. Verify all six configurations are available
# ------------------------------------------------

assert len(all_matrices) == 6

# Use a common maximum for comparable colour intensity
global_max = max(
    matrix.max() for matrix in all_matrices.values()
)

# ------------------------------------------------
# 5. Create the 2 x 3 publication-quality figure
# ------------------------------------------------

fig, axes = plt.subplots(
    nrows=2,
    ncols=3,
    figsize=(18, 11)
)

axes = axes.flatten()

panel_labels = ["(a)", "(b)", "(c)", "(d)", "(e)", "(f)"]

for index, (model_name, cm) in enumerate(all_matrices.items()):

    ax = axes[index]

    # Row-normalized percentages for interpretation
    row_totals = cm.sum(axis=1, keepdims=True)

    percentages = np.divide(
        cm,
        row_totals,
        out=np.zeros_like(cm, dtype=float),
        where=row_totals != 0
    ) * 100

    # Plot with a common colour scale
    image = ax.imshow(
        cm,
        interpolation="nearest",
        cmap="Blues",
        vmin=0,
        vmax=global_max
    )

    # Panel title
    ax.set_title(
        f"{panel_labels[index]} {model_name}",
        fontsize=13,
        fontweight="bold",
        pad=12
    )

    # Axis labels and class names
    ax.set_xticks(np.arange(len(class_names)))
    ax.set_yticks(np.arange(len(class_names)))

    ax.set_xticklabels(
        class_names,
        fontsize=11,
        rotation=0
    )

    ax.set_yticklabels(
        class_names,
        fontsize=11
    )

    ax.set_xlabel(
        "Predicted Class",
        fontsize=12,
        fontweight="bold",
        labelpad=8
    )

    ax.set_ylabel(
        "Actual Class",
        fontsize=12,
        fontweight="bold",
        labelpad=8
    )

    # Annotate each cell with count and row percentage
    threshold = global_max / 2

    for row in range(len(class_names)):
        for col in range(len(class_names)):

            count = cm[row, col]
            percentage = percentages[row, col]

            text_color = (
                "white" if count > threshold else "black"
            )

            ax.text(
                col,
                row,
                f"{count}\n({percentage:.1f}%)",
                ha="center",
                va="center",
                fontsize=11,
                fontweight="bold",
                color=text_color
            )

    # Improve panel appearance
    ax.set_xticks(
        np.arange(-0.5, len(class_names), 1),
        minor=True
    )

    ax.set_yticks(
        np.arange(-0.5, len(class_names), 1),
        minor=True
    )

    ax.grid(
        which="minor",
        color="white",
        linestyle="-",
        linewidth=2
    )

    ax.tick_params(
        which="minor",
        bottom=False,
        left=False
    )

# ------------------------------------------------
# 6. Add one shared colour bar
# ------------------------------------------------

cbar = fig.colorbar(
    image,
    ax=axes.tolist(),
    fraction=0.025,
    pad=0.025
)

cbar.set_label(
    "Number of Images",
    fontsize=12,
    fontweight="bold"
)

cbar.ax.tick_params(labelsize=10)

# ------------------------------------------------
# 7. Add overall figure title and layout
# ------------------------------------------------

fig.suptitle(
    "Confusion Matrices of Individual Backbone Models",
    fontsize=17,
    fontweight="bold",
    y=0.99
)

plt.subplots_adjust(
    left=0.07,
    right=0.91,
    top=0.91,
    bottom=0.08,
    wspace=0.30,
    hspace=0.32
)

# ------------------------------------------------
# 8. Save high-resolution PNG and vector PDF
# ------------------------------------------------

png_path = os.path.join(
    OUTPUT_DIR,
    "Figure2_Six_Individual_Backbone_Confusion_Matrices.png"
)

pdf_path = os.path.join(
    OUTPUT_DIR,
    "Figure2_Six_Individual_Backbone_Confusion_Matrices.pdf"
)

fig.savefig(
    png_path,
    dpi=600,
    bbox_inches="tight"
)

fig.savefig(
    pdf_path,
    format="pdf",
    bbox_inches="tight"
)

plt.show()

plt.close(fig)

print("\n================================================")
print("FIGURE 2 GENERATED")
print("================================================")
print("PNG:", png_path)
print("PDF:", pdf_path)
print("All six models contain 237 pooled predictions.")
```

```python

# =====================================================================
# FIGURE 2: CONFUSION MATRICES OF SIX INDIVIDUAL BACKBONE CONFIGURATIONS
# SCI JOURNAL PRESENTATION
#
# Source: Existing pooled out-of-fold confusion-matrix CSV files
# Dataset: 237 SEM images
# Classes: Bead, Fiber, Fragment
# Layout: 2 x 3
# Outputs: PNG (600 DPI), TIFF (600 DPI), PDF (vector)
# =====================================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ---------------------------------------------------------------------
# 1. Define directories
# ---------------------------------------------------------------------

BASE_DIR = "/content/drive/MyDrive/Microplastics_SEM"

CM_DIR = os.path.join(
    BASE_DIR,
    "AMAFFNet_Visualization",
    "Confusion_Matrices"
)

OUTPUT_DIR = CM_DIR

os.makedirs(OUTPUT_DIR, exist_ok=True)

# ---------------------------------------------------------------------
# 2. Define the six individual backbone configurations
# ---------------------------------------------------------------------

models = [
    "Xception_NoSE",
    "Xception_SE",
    "DenseNet201_NoSE",
    "DenseNet201_SE",
    "MobileNetV2_NoSE",
    "MobileNetV2_SE"
]

# Fixed class order throughout the figure
classes = ["Bead", "Fiber", "Fragment"]

# Expected class support in the complete dataset
expected_support = np.array([51, 90, 96])

expected_total = 237

# ---------------------------------------------------------------------
# 3. Load and validate the existing pooled confusion matrices
# ---------------------------------------------------------------------

confusion_matrices = {}

for model in models:

    file_name = f"{model}_Confusion_Matrix.csv"

    file_path = os.path.join(CM_DIR, file_name)

    if not os.path.exists(file_path):
        raise FileNotFoundError(
            f"Missing confusion matrix: {file_path}"
        )

    # Read the previously saved confusion matrix
    df = pd.read_csv(file_path)

    # The first column contains the actual class labels
    if "Unnamed: 0" not in df.columns:
        raise ValueError(
            f"Unexpected CSV structure for {model}"
        )

    df = df.set_index("Unnamed: 0")

    # Enforce the same row and column order
    df = df.reindex(index=classes, columns=classes)

    if df.isna().any().any():
        raise ValueError(
            f"Missing class values in {model}"
        )

    cm = df.to_numpy(dtype=int)

    # Verify that all 237 images are represented
    row_support = cm.sum(axis=1)

    if not np.array_equal(row_support, expected_support):
        raise ValueError(
            f"Class support mismatch for {model}: "
            f"{row_support.tolist()}"
        )

    if cm.sum() != expected_total:
        raise ValueError(
            f"Expected 237 predictions for {model}, "
            f"found {cm.sum()}"
        )

    confusion_matrices[model] = cm

    print(
        f"{model}: VALIDATED | "
        f"Total={cm.sum()} | "
        f"Class support={row_support.tolist()}"
    )

print("\nAll six pooled confusion matrices passed validation.")

# ---------------------------------------------------------------------
# 4. Establish a common colour scale
# ---------------------------------------------------------------------

# Identical colour limits allow visual comparison across panels
global_max = max(
    cm.max() for cm in confusion_matrices.values()
)

# ---------------------------------------------------------------------
# 5. Configure publication-quality plotting
# ---------------------------------------------------------------------

plt.rcParams.update({
    "font.family": "serif",
    "font.size": 11,
    "axes.titlesize": 13,
    "axes.labelsize": 11,
    "xtick.labelsize": 10,
    "ytick.labelsize": 10,
    "figure.titlesize": 16,
    "pdf.fonttype": 42,
    "ps.fonttype": 42
})

fig, axes = plt.subplots(
    2, 3,
    figsize=(17, 10),
    constrained_layout=False
)

axes = axes.flatten()

panel_labels = ["(a)", "(b)", "(c)", "(d)", "(e)", "(f)"]

# ---------------------------------------------------------------------
# 6. Plot the six confusion matrices
# ---------------------------------------------------------------------

for i, model in enumerate(models):

    ax = axes[i]

    cm = confusion_matrices[model]

    # Row-wise normalization for class-wise percentages
    row_totals = cm.sum(axis=1, keepdims=True)

    percentages = (
        cm / row_totals * 100
    )

    # Common blue colour scale
    im = ax.imshow(
        cm,
        cmap="Blues",
        vmin=0,
        vmax=global_max,
        aspect="equal",
        interpolation="nearest"
    )

    # Model name and panel designation
    ax.set_title(
        f"{panel_labels[i]} {model}",
        fontweight="bold",
        pad=12
    )

    # Class tick labels
    ax.set_xticks(range(3))
    ax.set_yticks(range(3))

    ax.set_xticklabels(classes)
    ax.set_yticklabels(classes)

    ax.set_xlabel("Predicted class", labelpad=7)
    ax.set_ylabel("Actual class", labelpad=7)

    # Cell annotations: count and row-normalized percentage
    for row in range(3):

        for col in range(3):

            count = cm[row, col]
            percentage = percentages[row, col]

            text_colour = (
                "white"
                if count > global_max * 0.50
                else "black"
            )

            ax.text(
                col,
                row,
                f"{count}\n{percentage:.1f}%",
                ha="center",
                va="center",
                color=text_colour,
                fontsize=11,
                fontweight="bold"
            )

    # Clear cell boundaries
    ax.set_xticks(
        np.arange(-0.5, 3, 1),
        minor=True
    )

    ax.set_yticks(
        np.arange(-0.5, 3, 1),
        minor=True
    )

    ax.grid(
        which="minor",
        color="white",
        linewidth=2
    )

    ax.tick_params(
        which="minor",
        bottom=False,
        left=False
    )

    # Subtle panel border
    for spine in ax.spines.values():
        spine.set_linewidth(0.8)

# ---------------------------------------------------------------------
# 7. Add one shared colour bar
# ---------------------------------------------------------------------

cbar = fig.colorbar(
    im,
    ax=axes.tolist(),
    shrink=0.88,
    pad=0.025,
    fraction=0.025
)

cbar.set_label(
    "Number of images",
    rotation=90,
    labelpad=12
)

# ---------------------------------------------------------------------
# 8. Add overall title and spacing
# ---------------------------------------------------------------------

fig.suptitle(
    "Confusion Matrices of Individual Backbone Configurations",
    fontweight="bold",
    y=0.98
)

fig.subplots_adjust(
    left=0.07,
    right=0.91,
    bottom=0.08,
    top=0.90,
    wspace=0.34,
    hspace=0.38
)

# ---------------------------------------------------------------------
# 9. Save publication-quality outputs
# ---------------------------------------------------------------------

png_path = os.path.join(
    OUTPUT_DIR,
    "Figure2_Individual_Backbone_Confusion_Matrices.png"
)

tiff_path = os.path.join(
    OUTPUT_DIR,
    "Figure2_Individual_Backbone_Confusion_Matrices.tiff"
)

pdf_path = os.path.join(
    OUTPUT_DIR,
    "Figure2_Individual_Backbone_Confusion_Matrices.pdf"
)

# High-resolution raster formats
fig.savefig(
    png_path,
    dpi=600,
    bbox_inches="tight"
)

fig.savefig(
    tiff_path,
    dpi=600,
    bbox_inches="tight",
    pil_kwargs={"compression": "tiff_lzw"}
)

# Vector format for manuscript preparation
fig.savefig(
    pdf_path,
    format="pdf",
    bbox_inches="tight"
)

plt.show()

plt.close(fig)

# ---------------------------------------------------------------------
# 10. Final output verification
# ---------------------------------------------------------------------

print("\n" + "=" * 70)
print("FIGURE 2 SUCCESSFULLY GENERATED")
print("=" * 70)

print("Models included:", len(confusion_matrices))
print("Predictions per model:", expected_total)
print("Class order:", classes)

print("\nSaved files:")

for path in [png_path, tiff_path, pdf_path]:
    print(path)
    print("File exists:", os.path.exists(path))
```

```python

# ================================================================
# FIGURE 3: COMPARATIVE ROC CURVES OF ALL EIGHT MODELS
# Uses pooled out-of-fold predictions from five-fold CV
# No model retraining and no numerical AUC labels in the figure
# ================================================================

import os
import glob
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import label_binarize
from sklearn.metrics import roc_curve, auc

# ------------------------------------------------
# 1. Define project directory and output directory
# ------------------------------------------------

BASE_DIR = (
    "/content/drive/MyDrive/Microplastics_SEM"
)

OUTPUT_DIR = os.path.join(
    BASE_DIR,
    "AMAFFNet_Visualization",
    "Comparison_Graphs"
)

os.makedirs(OUTPUT_DIR, exist_ok=True)

# ------------------------------------------------
# 2. Define the eight model configurations
# ------------------------------------------------

models = [
    "Xception_NoSE",
    "Xception_SE",
    "DenseNet201_NoSE",
    "DenseNet201_SE",
    "MobileNetV2_NoSE",
    "MobileNetV2_SE",
    "AMAFFNet_NoSE",
    "AMAFFNet_SE"
]

# Fixed class order for all models
classes = ["Bead", "Fiber", "Fragment"]

# Probability column alternatives found in saved prediction files
probability_aliases = {
    "Bead": [
        "Bead_Probability",
        "Probability_Bead",
        "Bead_Prob",
        "prob_bead"
    ],
    "Fiber": [
        "Fiber_Probability",
        "Probability_Fiber",
        "Fiber_Prob",
        "prob_fiber"
    ],
    "Fragment": [
        "Fragment_Probability",
        "Probability_Fragment",
        "Fragment_Prob",
        "prob_fragment"
    ]
}

# ------------------------------------------------
# 3. Function to locate pooled prediction files
# ------------------------------------------------

def locate_prediction_file(model):

    # Search for saved all-fold prediction files
    patterns = [
        os.path.join(
            BASE_DIR, "**",
            f"{model}_All_Fold_Predictions.csv"
        ),
        os.path.join(
            BASE_DIR, "**",
            f"*{model}*All_Fold_Predictions.csv"
        )
    ]

    candidates = []

    for pattern in patterns:
        candidates.extend(
            glob.glob(pattern, recursive=True)
        )

    # Remove duplicate paths
    candidates = list(dict.fromkeys(candidates))

    if len(candidates) == 0:
        raise FileNotFoundError(
            f"No pooled prediction CSV found for {model}."
        )

    # Prefer a file containing all 237 pooled predictions
    valid_candidates = []

    for path in candidates:
        try:
            temp = pd.read_csv(path)

            if len(temp) == 237:
                valid_candidates.append(path)

        except Exception:
            continue

    if len(valid_candidates) == 0:
        raise ValueError(
            f"No 237-row pooled prediction file found for {model}. "
            f"Candidates: {candidates}"
        )

    if len(valid_candidates) > 1:
        print(f"\nMultiple pooled files found for {model}:")
        for path in valid_candidates:
            print(path)

        # Prefer the file under the main results directory
        preferred = [
            p for p in valid_candidates
            if "AMAFFNet_Results" in p
        ]

        if len(preferred) == 1:
            return preferred[0]

        raise ValueError(
            f"Please identify the correct pooled prediction file "
            f"for {model}; multiple valid files were found."
        )

    return valid_candidates[0]


# ------------------------------------------------
# 4. Function to standardize prediction columns
# ------------------------------------------------

def prepare_prediction_data(model):

    file_path = locate_prediction_file(model)

    print(f"\nLoading: {model}")
    print("File:", file_path)

    df = pd.read_csv(file_path)

    # Detect the actual class column
    actual_options = [
        "Actual_Class",
        "True_Class",
        "Actual",
        "True_Label",
        "Ground_Truth"
    ]

    actual_col = next(
        (
            col for col in actual_options
            if col in df.columns
        ),
        None
    )

    if actual_col is None:
        raise ValueError(
            f"Actual class column not found for {model}. "
            f"Available columns: {list(df.columns)}"
        )

    # Detect the three class-probability columns
    selected_probability_columns = {}

    for class_name in classes:

        found_col = next(
            (
                col for col in probability_aliases[class_name]
                if col in df.columns
            ),
            None
        )

        if found_col is None:
            raise ValueError(
                f"Probability column for {class_name} "
                f"not found in {model}. "
                f"Available columns: {list(df.columns)}"
            )

        selected_probability_columns[class_name] = found_col

    # Extract actual labels and predicted probabilities
    y_true = df[actual_col].astype(str).str.strip().to_numpy()

    probability_columns = [
        selected_probability_columns[c]
        for c in classes
    ]

    y_prob = df[probability_columns].to_numpy(dtype=float)

    # Check for missing or invalid probability values
    if not np.isfinite(y_prob).all():
        raise ValueError(
            f"Missing or invalid probability values in {model}"
        )

    if np.any(y_prob < 0) or np.any(y_prob > 1):
        raise ValueError(
            f"Probabilities outside [0,1] in {model}"
        )

    # Verify probability sums
    if not np.allclose(
        y_prob.sum(axis=1),
        1.0,
        atol=1e-4
    ):
        raise ValueError(
            f"Probability sums are not equal to 1 for {model}"
        )

    # Verify the three expected classes
    if set(np.unique(y_true)) != set(classes):
        raise ValueError(
            f"Unexpected class labels in {model}: "
            f"{np.unique(y_true)}"
        )

    # Verify class support
    actual_counts = pd.Series(y_true).value_counts()

    expected_counts = {
        "Bead": 51,
        "Fiber": 90,
        "Fragment": 96
    }

    for class_name, expected in expected_counts.items():

        if actual_counts.get(class_name, 0) != expected:
            raise ValueError(
                f"Incorrect support for {class_name} in {model}"
            )

    print("Total predictions:", len(y_true))
    print("Class support:", actual_counts.to_dict())
    print("Probability verification: PASSED")

    return y_true, y_prob


# ------------------------------------------------
# 5. Load all eight pooled prediction files
# ------------------------------------------------

model_predictions = {}

for model in models:

    y_true, y_prob = prepare_prediction_data(model)

    model_predictions[model] = (y_true, y_prob)

print("\nAll eight prediction files loaded successfully.")


# ------------------------------------------------
# 6. Define ROC calculation
# ------------------------------------------------

def calculate_macro_roc(y_true, y_prob):

    # Convert actual class labels into one-vs-rest format
    y_binary = label_binarize(
        y_true,
        classes=classes
    )

    # Common false-positive-rate points for averaging
    mean_fpr = np.linspace(0, 1, 1000)

    interpolated_tprs = []
    individual_aucs = []

    # Calculate one ROC curve per class
    for i in range(len(classes)):

        fpr, tpr, _ = roc_curve(
            y_binary[:, i],
            y_prob[:, i]
        )

        class_auc = auc(fpr, tpr)
        individual_aucs.append(class_auc)

        # Interpolate each class curve
        interp_tpr = np.interp(
            mean_fpr,
            fpr,
            tpr
        )

        interp_tpr[0] = 0.0

        interpolated_tprs.append(interp_tpr)

    # Calculate macro-average ROC curve
    mean_tpr = np.mean(
        interpolated_tprs,
        axis=0
    )

    mean_tpr[-1] = 1.0

    macro_auc = auc(
        mean_fpr,
        mean_tpr
    )

    return mean_fpr, mean_tpr, macro_auc


# ------------------------------------------------
# 7. Plot all eight macro-average ROC curves
# ------------------------------------------------

plt.rcParams["font.family"] = "DejaVu Sans"

fig, ax = plt.subplots(
    figsize=(12, 10)
)

# Distinct colours for all eight configurations
colors = [
    "#4472C4",
    "#70AD47",
    "#ED7D31",
    "#FFC000",
    "#8064A2",
    "#4BACC6",
    "#C0504D",
    "#17365D"
]

line_styles = [
    "-",
    "--",
    "-.",
    ":",
    "-",
    "--",
    "-.",
    ":"
]

for i, model in enumerate(models):

    y_true, y_prob = model_predictions[model]

    fpr, tpr, macro_auc = calculate_macro_roc(
        y_true,
        y_prob
    )

    # Plot without numeric AUC values in the legend
    ax.plot(
        fpr,
        tpr,
        color=colors[i],
        linestyle=line_styles[i],
        linewidth=2.5,
        label=model.replace("_", " ")
    )

# ------------------------------------------------
# 8. Add random-classifier reference line
# ------------------------------------------------

ax.plot(
    [0, 1],
    [0, 1],
    color="gray",
    linestyle="--",
    linewidth=1.5,
    label="Random Classifier"
)

# ------------------------------------------------
# 9. Format the ROC graph
# ------------------------------------------------

ax.set_xlim(0, 1)
ax.set_ylim(0, 1.02)

ax.set_xlabel(
    "False Positive Rate",
    fontsize=14,
    fontweight="bold",
    labelpad=10
)

ax.set_ylabel(
    "True Positive Rate",
    fontsize=14,
    fontweight="bold",
    labelpad=10
)

ax.set_title(
    "Comparative ROC Curves of Eight Model Configurations",
    fontsize=16,
    fontweight="bold",
    pad=18
)

ax.tick_params(
    axis="both",
    labelsize=11
)

ax.grid(
    True,
    linestyle="--",
    linewidth=0.6,
    alpha=0.35
)

# Place legend outside the plotting area
ax.legend(
    loc="center left",
    bbox_to_anchor=(1.02, 0.5),
    fontsize=10,
    frameon=True,
    edgecolor="gray"
)

# ------------------------------------------------
# 10. Save publication-quality figures
# ------------------------------------------------

fig.tight_layout()

fig.savefig(
    os.path.join(
        OUTPUT_DIR,
        "Figure3_Comparative_ROC_All_Eight_Models.png"
    ),
    dpi=600,
    bbox_inches="tight"
)

fig.savefig(
    os.path.join(
        OUTPUT_DIR,
        "Figure3_Comparative_ROC_All_Eight_Models.tiff"
    ),
    dpi=600,
    bbox_inches="tight",
    pil_kwargs={"compression": "tiff_lzw"}
)

fig.savefig(
    os.path.join(
        OUTPUT_DIR,
        "Figure3_Comparative_ROC_All_Eight_Models.pdf"
    ),
    bbox_inches="tight"
)

plt.show()
plt.close(fig)

print("\nFigure 3 generated successfully.")
print("Saved in:", OUTPUT_DIR)
```

```python

# ================================================================
# CONSOLIDATE AMAFFNET NoSE AND SE OUT-OF-FOLD PREDICTIONS
# Combines the five saved test-fold prediction files
# Validates sample counts, folds, and unique image paths
# ================================================================

import os
import pandas as pd

# ------------------------------------------------
# 1. DEFINE DIRECTORIES
# ------------------------------------------------

BASE_DIR = "/content/drive/MyDrive/Microplastics_SEM"

RESULT_DIR = os.path.join(
    BASE_DIR, "AMAFFNet_Fusion_Results"
)

OUTPUT_DIR = os.path.join(
    BASE_DIR, "AMAFFNet_Visualization", "ROC_Curves"
)

os.makedirs(OUTPUT_DIR, exist_ok=True)

EXPERIMENTS = [
    "AMAFFNet_NoSE",
    "AMAFFNet_SE"
]

EXPECTED_SAMPLES = 237
EXPECTED_FOLDS = {1, 2, 3, 4, 5}

# ------------------------------------------------
# 2. COMBINE FIVE FOLD PREDICTION FILES
# ------------------------------------------------

for experiment in EXPERIMENTS:

    print("\n" + "=" * 80)
    print("PROCESSING:", experiment)
    print("=" * 80)

    experiment_dir = os.path.join(
        RESULT_DIR, experiment
    )

    fold_dataframes = []

    # Read the five saved outer-test-fold prediction files
    for fold in range(1, 6):

        prediction_file = os.path.join(
            experiment_dir,
            f"Fold_{fold}",
            "Test_Predictions.csv"
        )

        completion_file = os.path.join(
            experiment_dir,
            f"Fold_{fold}",
            "FOLD_COMPLETE.json"
        )

        # Verify that the fold was completed
        if not os.path.exists(completion_file):
            raise FileNotFoundError(
                f"Fold completion marker missing: {completion_file}"
            )

        # Verify that predictions were saved
        if not os.path.exists(prediction_file):
            raise FileNotFoundError(
                f"Prediction file missing: {prediction_file}"
            )

        df = pd.read_csv(prediction_file)

        # Add fold identifier to the prediction records
        df["Fold"] = fold

        fold_dataframes.append(df)

        print(
            f"Fold {fold}: {len(df)} predictions loaded."
        )

    # ------------------------------------------------
    # 3. CONCATENATE ALL FIVE FOLDS
    # ------------------------------------------------

    combined_df = pd.concat(
        fold_dataframes,
        ignore_index=True
    )

    # ------------------------------------------------
    # 4. VERIFY REQUIRED COLUMNS
    # ------------------------------------------------

    required_columns = [
        "Image_Path",
        "True_Label",
        "True_Class",
        "Predicted_Label",
        "Predicted_Class",
        "Probability_Bead",
        "Probability_Fiber",
        "Probability_Fragment",
        "Fold"
    ]

    missing_columns = [
        col for col in required_columns
        if col not in combined_df.columns
    ]

    if missing_columns:
        raise ValueError(
            f"Missing columns: {missing_columns}"
        )

    # ------------------------------------------------
    # 5. VERIFY FIVE-FOLD COMPLETENESS
    # ------------------------------------------------

    assert len(combined_df) == EXPECTED_SAMPLES, (
        f"Expected {EXPECTED_SAMPLES} predictions, "
        f"found {len(combined_df)}."
    )

    assert set(combined_df["Fold"].unique()) == EXPECTED_FOLDS

    # Each image must occur in exactly one outer test fold
    assert not combined_df["Image_Path"].duplicated().any(), (
        "Duplicate image paths detected across test folds."
    )

    # Verify class probability values
    probability_columns = [
        "Probability_Bead",
        "Probability_Fiber",
        "Probability_Fragment"
    ]

    assert combined_df[probability_columns].notna().all().all()

    # Each probability row should sum approximately to one
    probability_sums = combined_df[
        probability_columns
    ].sum(axis=1)

    assert ((probability_sums - 1.0).abs() < 1e-4).all(), (
        "Some probability rows do not sum to approximately one."
    )

    # ------------------------------------------------
    # 6. SAVE CONSOLIDATED PREDICTIONS
    # ------------------------------------------------

    output_file = os.path.join(
        OUTPUT_DIR,
        f"{experiment}_All_Fold_Predictions.csv"
    )

    temporary_file = output_file + ".tmp"

    combined_df.to_csv(
        temporary_file,
        index=False
    )

    os.replace(
        temporary_file,
        output_file
    )

    # ------------------------------------------------
    # 7. DISPLAY VERIFICATION SUMMARY
    # ------------------------------------------------

    print("\nVALIDATION PASSED")
    print("Total predictions:", len(combined_df))
    print("Unique images:", combined_df["Image_Path"].nunique())
    print("Folds:", sorted(combined_df["Fold"].unique()))

    print("\nClass distribution:")
    print(combined_df["True_Class"].value_counts())

    print("\nSaved consolidated file:")
    print(output_file)

print("\n" + "=" * 80)
print("BOTH AMAFFNET VARIANTS CONSOLIDATED SUCCESSFULLY")
print("=" * 80)
```

```python

# ================================================================
# FINAL VERIFICATION OF INDIVIDUAL BACKBONE PREDICTION FILES
# Checks NoSE and SE prediction sources before ROC plotting
# No model training or modification of existing files
# ================================================================

import os
import numpy as np
import pandas as pd

BASE = "/content/drive/MyDrive/Microplastics_SEM"

RESULTS = os.path.join(
    BASE, "AMAFFNet_Results"
)

VIS = os.path.join(
    BASE, "AMAFFNet_Visualization"
)

# ------------------------------------------------
# 1. DEFINE THE EXACT FILE PATHS
# ------------------------------------------------

files = {
    "Xception_Generic": os.path.join(
        RESULTS, "Xception",
        "Xception_All_Fold_Predictions.csv"
    ),

    "DenseNet201_Generic": os.path.join(
        RESULTS, "DenseNet201",
        "DenseNet201_All_Fold_Predictions.csv"
    ),

    "DenseNet201_NoSE": os.path.join(
        RESULTS, "DenseNet201_NoSE",
        "DenseNet201_NoSE_All_Fold_Predictions.csv"
    ),

    "MobileNetV2_Generic": os.path.join(
        RESULTS, "MobileNetV2",
        "MobileNetV2_All_Fold_Predictions.csv"
    ),

    "MobileNetV2_NoSE": os.path.join(
        RESULTS, "MobileNetV2_NoSE",
        "MobileNetV2_NoSE_All_Fold_Predictions.csv"
    ),

    "Xception_SE": os.path.join(
        VIS, "Xception_SE_SE_Standardized_Predictions.csv"
    ),

    "DenseNet201_SE": os.path.join(
        VIS, "DenseNet201_SE_SE_Standardized_Predictions.csv"
    ),

    "MobileNetV2_SE": os.path.join(
        VIS, "MobileNetV2_SE_SE_Standardized_Predictions.csv"
    )
}

# ------------------------------------------------
# 2. READ AND INSPECT EACH FILE
# ------------------------------------------------

loaded = {}

for name, path in files.items():

    print("\n" + "=" * 85)
    print("MODEL:", name)
    print("=" * 85)

    if not os.path.exists(path):
        print("FILE NOT FOUND:", path)
        continue

    df = pd.read_csv(path)
    loaded[name] = df

    print("Rows:", len(df))
    print("Unique image paths:",
          df["Image_Path"].nunique()
          if "Image_Path" in df.columns else "Image_Path missing")

    print("Columns:", df.columns.tolist())

    # Identify probability columns
    probability_columns = [
        col for col in df.columns
        if "probability" in col.lower()
    ]

    print("Probability columns:", probability_columns)

    # Identify the actual class column
    actual_candidates = [
        "True_Class", "Actual_Class", "Class"
    ]

    actual_column = next(
        (col for col in actual_candidates if col in df.columns),
        None
    )

    if actual_column:
        print("\nClass distribution:")
        print(df[actual_column].value_counts().to_dict())

    if len(df) != 237:
        print("WARNING: Expected 237 rows.")

    if "Image_Path" in df.columns:
        if df["Image_Path"].duplicated().any():
            print("WARNING: Duplicate image paths detected.")

    if len(probability_columns) == 3:
        sums = df[probability_columns].sum(axis=1)
        print(
            "Probability sum range:",
            round(sums.min(), 6),
            "to",
            round(sums.max(), 6)
        )

# ------------------------------------------------
# 3. COMPARE GENERIC AND EXPLICIT NoSE PREDICTIONS
# ------------------------------------------------

comparisons = [
    ("DenseNet201_Generic", "DenseNet201_NoSE"),
    ("MobileNetV2_Generic", "MobileNetV2_NoSE")
]

print("\n" + "=" * 85)
print("GENERIC VERSUS EXPLICIT NoSE PREDICTION COMPARISON")
print("=" * 85)

for first, second in comparisons:

    if first not in loaded or second not in loaded:
        continue

    df1 = loaded[first].copy()
    df2 = loaded[second].copy()

    probability_cols_1 = [
        c for c in df1.columns
        if "probability" in c.lower()
    ]

    probability_cols_2 = [
        c for c in df2.columns
        if "probability" in c.lower()
    ]

    if len(probability_cols_1) != 3 or len(probability_cols_2) != 3:
        print(f"\n{first} vs {second}: Cannot compare probability columns.")
        continue

    # Standardize probability column order by class name
    def get_probability_columns(df):
        mapping = {}

        for col in df.columns:
            lower = col.lower()

            if "probability" in lower:
                if "bead" in lower:
                    mapping["Bead"] = col
                elif "fiber" in lower:
                    mapping["Fiber"] = col
                elif "fragment" in lower:
                    mapping["Fragment"] = col

        return mapping

    map1 = get_probability_columns(df1)
    map2 = get_probability_columns(df2)

    if set(map1) != {"Bead", "Fiber", "Fragment"} or \
       set(map2) != {"Bead", "Fiber", "Fragment"}:
        print(f"\n{first} vs {second}: Class probability mapping incomplete.")
        continue

    # Match predictions by image path
    merged = df1[["Image_Path"]].copy()

    for cls in ["Bead", "Fiber", "Fragment"]:
        merged[cls + "_A"] = df1.set_index(
            "Image_Path"
        )[map1[cls]].reindex(merged["Image_Path"]).values

        merged[cls + "_B"] = df2.set_index(
            "Image_Path"
        )[map2[cls]].reindex(merged["Image_Path"]).values

    if len(merged) != 237 or merged.isna().any().any():
        print(f"\n{first} vs {second}: Image matching incomplete.")
        continue

    differences = []

    for cls in ["Bead", "Fiber", "Fragment"]:
        differences.extend(
            np.abs(
                merged[cls + "_A"].to_numpy() -
                merged[cls + "_B"].to_numpy()
            )
        )

    max_difference = float(np.max(differences))
    mean_difference = float(np.mean(differences))

    print(f"\n{first} vs {second}")
    print("Matched images:", len(merged))
    print("Maximum probability difference:", max_difference)
    print("Mean probability difference:", mean_difference)
    print("Predictions identical within 1e-7:",
          np.allclose(differences, 0, atol=1e-7))

print("\n" + "=" * 85)
print("INSPECTION COMPLETED")
print("=" * 85)
```

```python

# ================================================================
# XCEPTION_NoSE: GENERATE POOLED FIVE-FOLD TEST PREDICTIONS
# Uses the five existing trained checkpoints
# No model retraining
# ================================================================

import os
import gc
import numpy as np
import pandas as pd
import tensorflow as tf

from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications.xception import preprocess_input
from tensorflow.keras.models import load_model

# ------------------------------------------------
# STEP 1: DEFINE PROJECT DIRECTORIES
# ------------------------------------------------

BASE_DIR = "/content/drive/MyDrive/Microplastics_SEM"

SPLIT_DIR = os.path.join(
    BASE_DIR, "AMAFFNet_5Fold_Splits"
)

MODEL_DIR = os.path.join(
    BASE_DIR, "AMAFFNet_Weights", "Xception"
)

OUTPUT_DIR = os.path.join(
    BASE_DIR, "AMAFFNet_Results", "Xception"
)

os.makedirs(OUTPUT_DIR, exist_ok=True)

OUTPUT_FILE = os.path.join(
    OUTPUT_DIR,
    "Xception_All_Fold_Predictions.csv"
)

# ------------------------------------------------
# STEP 2: CONFIGURATION
# ------------------------------------------------

IMG_SIZE = (299, 299)
BATCH_SIZE = 16

CLASS_NAMES = [
    "Bead",
    "Fiber",
    "Fragment"
]

# Preprocessing only; no augmentation during testing
test_datagen = ImageDataGenerator(
    preprocessing_function=preprocess_input
)

all_fold_predictions = []

# ------------------------------------------------
# STEP 3: PROCESS ALL FIVE FOLDS
# ------------------------------------------------

for fold in range(1, 6):

    print("\n" + "=" * 75)
    print(f"GENERATING XCEPTION_NoSE PREDICTIONS - FOLD {fold}")
    print("=" * 75)

    # Clear previous fold resources
    tf.keras.backend.clear_session()
    gc.collect()

    # ------------------------------------------------
    # Load the original held-out test split
    # ------------------------------------------------

    test_path = os.path.join(
        SPLIT_DIR,
        f"fold_{fold}_test.csv"
    )

    test_df = pd.read_csv(test_path).reset_index(drop=True)

    # ------------------------------------------------
    # Load the corresponding trained Xception checkpoint
    # ------------------------------------------------

    checkpoint_path = os.path.join(
        MODEL_DIR,
        f"best_xception_fold_{fold}.keras"
    )

    if not os.path.isfile(checkpoint_path):
        raise FileNotFoundError(
            f"Missing checkpoint: {checkpoint_path}"
        )

    model = load_model(checkpoint_path)

    print("Checkpoint loaded:", checkpoint_path)

    # ------------------------------------------------
    # Create test generator
    # shuffle=False preserves image order
    # ------------------------------------------------

    test_generator = test_datagen.flow_from_dataframe(
        dataframe=test_df,
        x_col="Image_Path",
        y_col="Class",
        target_size=IMG_SIZE,
        batch_size=BATCH_SIZE,
        class_mode="categorical",
        classes=CLASS_NAMES,
        shuffle=False
    )

    # ------------------------------------------------
    # Predict class probabilities
    # ------------------------------------------------

    probabilities = model.predict(
        test_generator,
        verbose=1
    )

    # Confirm prediction dimensions
    if probabilities.shape != (len(test_df), 3):
        raise ValueError(
            f"Unexpected prediction shape in fold {fold}: "
            f"{probabilities.shape}"
        )

    # ------------------------------------------------
    # Convert probabilities to predicted class labels
    # ------------------------------------------------

    predicted_indices = np.argmax(
        probabilities,
        axis=1
    )

    predicted_classes = [
        CLASS_NAMES[index]
        for index in predicted_indices
    ]

    true_classes = test_df["Class"].astype(str).values

    # ------------------------------------------------
    # Create fold prediction dataframe
    # ------------------------------------------------

    fold_predictions = pd.DataFrame({

        "Fold": fold,

        "Image_Path": test_df["Image_Path"].values,

        "True_Class": true_classes,

        "Predicted_Class": predicted_classes,

        "Probability_Bead": probabilities[:, 0],

        "Probability_Fiber": probabilities[:, 1],

        "Probability_Fragment": probabilities[:, 2],

        "Correct": (
            true_classes == np.array(predicted_classes)
        ).astype(int)
    })

    # Add completed fold predictions to the collection
    all_fold_predictions.append(fold_predictions)

    print("Fold:", fold)
    print("Test images:", len(fold_predictions))
    print(
        "Fold accuracy:",
        round(fold_predictions["Correct"].mean(), 4)
    )

    # Release model and generator resources
    del model, test_generator
    gc.collect()
    tf.keras.backend.clear_session()

# ------------------------------------------------
# STEP 4: COMBINE ALL FIVE FOLDS
# ------------------------------------------------

final_predictions = pd.concat(
    all_fold_predictions,
    ignore_index=True
)

# ------------------------------------------------
# STEP 5: VALIDATION
# ------------------------------------------------

print("\n" + "=" * 75)
print("FINAL POOLED PREDICTION VALIDATION")
print("=" * 75)

print("Total predictions:", len(final_predictions))

print(
    "Unique images:",
    final_predictions["Image_Path"].nunique()
)

print(
    "Fold distribution:",
    final_predictions["Fold"].value_counts().sort_index().to_dict()
)

print("\nClass distribution:")
print(final_predictions["True_Class"].value_counts())

# Check expected dataset size
assert len(final_predictions) == 237, \
    "Expected exactly 237 predictions."

assert final_predictions["Image_Path"].nunique() == 237, \
    "Duplicate image paths detected."

assert set(final_predictions["Fold"].unique()) == {
    1, 2, 3, 4, 5
}, "Unexpected fold numbers."

# Check for missing values
assert not final_predictions.isnull().any().any(), \
    "Missing values detected."

# Check class probability sums
probability_columns = [
    "Probability_Bead",
    "Probability_Fiber",
    "Probability_Fragment"
]

probability_sums = final_predictions[
    probability_columns
].sum(axis=1)

assert np.allclose(
    probability_sums,
    1.0,
    atol=1e-5
), "Probability sums are not equal to 1."

# Check expected class counts
expected_counts = {
    "Bead": 51,
    "Fiber": 90,
    "Fragment": 96
}

actual_counts = final_predictions[
    "True_Class"
].value_counts().to_dict()

assert actual_counts == expected_counts, \
    f"Unexpected class distribution: {actual_counts}"

print("\nALL VALIDATION CHECKS PASSED")

# ------------------------------------------------
# STEP 6: SAVE POOLED PREDICTIONS
# ------------------------------------------------

final_predictions.to_csv(
    OUTPUT_FILE,
    index=False
)

print("\nSaved prediction file:")
print(OUTPUT_FILE)

# ------------------------------------------------
# STEP 7: RELOAD AND VERIFY SAVED FILE
# ------------------------------------------------

verified_df = pd.read_csv(OUTPUT_FILE)

assert len(verified_df) == 237
assert verified_df["Image_Path"].nunique() == 237

print("\nSaved file verification: PASSED")
print("Final rows:", len(verified_df))
print("Final columns:", verified_df.columns.tolist())

print("\n" + "=" * 75)
print("XCEPTION_NoSE PREDICTION GENERATION COMPLETED")
print("=" * 75)
```

```python

# ================================================================
# FIGURE 3: COMPARATIVE ROC CURVES OF ALL EIGHT MODEL CONFIGURATIONS
#
# Dataset: Mendeley SEM - 237 original images
# Evaluation: Stratified five-fold cross-validation
# ROC: One-vs-rest macro-average using pooled OOF probabilities
#
# No model retraining
# No numerical AUC labels in the figure
# No percentages or accuracy values in the figure
# ================================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import label_binarize
from sklearn.metrics import roc_curve, auc

# ------------------------------------------------
# STEP 1: DEFINE DIRECTORIES
# ------------------------------------------------

BASE_DIR = "/content/drive/MyDrive/Microplastics_SEM"

RESULTS_DIR = os.path.join(
    BASE_DIR, "AMAFFNet_Results"
)

VIS_DIR = os.path.join(
    BASE_DIR, "AMAFFNet_Visualization"
)

ROC_DIR = os.path.join(
    VIS_DIR, "ROC_Curves"
)

GRAPH_DIR = os.path.join(
    VIS_DIR, "Comparison_Graphs"
)

os.makedirs(GRAPH_DIR, exist_ok=True)

# ------------------------------------------------
# STEP 2: DEFINE CLASS ORDER
# ------------------------------------------------

CLASS_NAMES = [
    "Bead",
    "Fiber",
    "Fragment"
]

EXPECTED_COUNTS = {
    "Bead": 51,
    "Fiber": 90,
    "Fragment": 96
}

# ------------------------------------------------
# STEP 3: DEFINE THE EIGHT VERIFIED PREDICTION FILES
# ------------------------------------------------

MODEL_FILES = {

    "Xception (without SE)": {
        "path": os.path.join(
            RESULTS_DIR,
            "Xception",
            "Xception_All_Fold_Predictions.csv"
        ),
        "actual": "True_Class",
        "probabilities": [
            "Probability_Bead",
            "Probability_Fiber",
            "Probability_Fragment"
        ]
    },

    "Xception (with SE)": {
        "path": os.path.join(
            VIS_DIR,
            "Xception_SE_SE_Standardized_Predictions.csv"
        ),
        "actual": "Actual",
        "probabilities": [
            "Prob_Bead",
            "Prob_Fiber",
            "Prob_Fragment"
        ]
    },

    "DenseNet201 (without SE)": {
        "path": os.path.join(
            RESULTS_DIR,
            "DenseNet201_NoSE",
            "DenseNet201_NoSE_All_Fold_Predictions.csv"
        ),
        "actual": "Actual_Class",
        "probabilities": [
            "Bead_Probability",
            "Fiber_Probability",
            "Fragment_Probability"
        ]
    },

    "DenseNet201 (with SE)": {
        "path": os.path.join(
            VIS_DIR,
            "DenseNet201_SE_SE_Standardized_Predictions.csv"
        ),
        "actual": "Actual",
        "probabilities": [
            "Prob_Bead",
            "Prob_Fiber",
            "Prob_Fragment"
        ]
    },

    "MobileNetV2 (without SE)": {
        "path": os.path.join(
            RESULTS_DIR,
            "MobileNetV2_NoSE",
            "MobileNetV2_NoSE_All_Fold_Predictions.csv"
        ),
        "actual": "True_Class",
        "probabilities": [
            "Probability_Bead",
            "Probability_Fiber",
            "Probability_Fragment"
        ]
    },

    "MobileNetV2 (with SE)": {
        "path": os.path.join(
            VIS_DIR,
            "MobileNetV2_SE_SE_Standardized_Predictions.csv"
        ),
        "actual": "Actual",
        "probabilities": [
            "Prob_Bead",
            "Prob_Fiber",
            "Prob_Fragment"
        ]
    },

    "AMAFFNet (without SE)": {
        "path": os.path.join(
            ROC_DIR,
            "AMAFFNet_NoSE_All_Fold_Predictions.csv"
        ),
        "actual": "True_Class",
        "probabilities": [
            "Probability_Bead",
            "Probability_Fiber",
            "Probability_Fragment"
        ]
    },

    "AMAFFNet (with SE)": {
        "path": os.path.join(
            ROC_DIR,
            "AMAFFNet_SE_All_Fold_Predictions.csv"
        ),
        "actual": "True_Class",
        "probabilities": [
            "Probability_Bead",
            "Probability_Fiber",
            "Probability_Fragment"
        ]
    }
}

# ------------------------------------------------
# STEP 4: LOAD AND VALIDATE PREDICTIONS
# ------------------------------------------------

model_data = {}

print("=" * 85)
print("VALIDATING EIGHT MODEL PREDICTION FILES")
print("=" * 85)

for model_name, config in MODEL_FILES.items():

    file_path = config["path"]

    print("\nChecking:", model_name)

    # Check whether the prediction file exists
    if not os.path.isfile(file_path):
        raise FileNotFoundError(
            f"Prediction file not found:\n{file_path}"
        )

    df = pd.read_csv(file_path)

    # Check the required columns
    required_columns = [
        config["actual"]
    ] + config["probabilities"]

    missing_columns = [
        col for col in required_columns
        if col not in df.columns
    ]

    if missing_columns:
        raise ValueError(
            f"{model_name}: Missing columns {missing_columns}\n"
            f"Available columns: {df.columns.tolist()}"
        )

    # Extract actual class labels
    y_true = df[config["actual"]].copy()

    # Support numeric labels if present
    if pd.api.types.is_numeric_dtype(y_true):
        label_mapping = {
            0: "Bead",
            1: "Fiber",
            2: "Fragment"
        }

        y_true = y_true.map(label_mapping)

    y_true = y_true.astype(str).str.strip()

    # Extract class probabilities in fixed class order
    y_prob = df[
        config["probabilities"]
    ].to_numpy(dtype=float)

    # Check the number of predictions
    if len(df) != 237:
        raise ValueError(
            f"{model_name}: Expected 237 rows, found {len(df)}"
        )

    # Check class labels
    if not set(y_true.unique()).issubset(set(CLASS_NAMES)):
        raise ValueError(
            f"{model_name}: Unexpected class labels "
            f"{y_true.unique().tolist()}"
        )

    # Check class distribution
    actual_counts = y_true.value_counts().to_dict()

    if actual_counts != EXPECTED_COUNTS:
        raise ValueError(
            f"{model_name}: Incorrect class distribution "
            f"{actual_counts}"
        )

    # Check probability values
    if not np.isfinite(y_prob).all():
        raise ValueError(
            f"{model_name}: Invalid probability values."
        )

    if np.any(y_prob < 0) or np.any(y_prob > 1):
        raise ValueError(
            f"{model_name}: Probabilities outside [0, 1]."
        )

    # Check that probabilities sum to approximately one
    if not np.allclose(
        y_prob.sum(axis=1),
        1.0,
        atol=1e-4
    ):
        raise ValueError(
            f"{model_name}: Probability sums are invalid."
        )

    # Convert actual labels into one-hot binary format
    y_binary = label_binarize(
        y_true,
        classes=CLASS_NAMES
    )

    model_data[model_name] = {
        "y_binary": y_binary,
        "y_prob": y_prob
    }

    print("Rows:", len(df))
    print("Class distribution:", actual_counts)
    print("Probability validation: PASSED")

print("\nALL EIGHT MODELS PASSED VALIDATION.")

# ------------------------------------------------
# STEP 5: COMPUTE MACRO-AVERAGE ROC CURVES
# ------------------------------------------------

# Common false-positive-rate points for interpolation
mean_fpr = np.linspace(0, 1, 1000)

roc_results = {}

for model_name, data in model_data.items():

    y_binary = data["y_binary"]
    y_prob = data["y_prob"]

    class_tprs = []
    class_aucs = []

    # Calculate one-vs-rest ROC for each class
    for class_index in range(3):

        fpr, tpr, thresholds = roc_curve(
            y_binary[:, class_index],
            y_prob[:, class_index]
        )

        # Interpolate each class ROC onto common FPR points
        interpolated_tpr = np.interp(
            mean_fpr,
            fpr,
            tpr
        )

        interpolated_tpr[0] = 0.0

        class_tprs.append(interpolated_tpr)

        class_aucs.append(
            auc(fpr, tpr)
        )

    # Average the three class-specific ROC curves
    macro_tpr = np.mean(
        class_tprs,
        axis=0
    )

    macro_tpr[0] = 0.0
    macro_tpr[-1] = 1.0

    # Calculate macro-average AUC internally
    macro_auc = auc(
        mean_fpr,
        macro_tpr
    )

    roc_results[model_name] = {
        "fpr": mean_fpr,
        "tpr": macro_tpr,
        "auc": macro_auc
    }

    print(
        f"{model_name}: Macro ROC-AUC = {macro_auc:.6f}"
    )

# ------------------------------------------------
# STEP 6: PLOT ALL EIGHT ROC CURVES
# ------------------------------------------------

# Create a publication-quality figure
fig, ax = plt.subplots(
    figsize=(11, 8),
    dpi=150
)

# Use eight visually distinct colours
colors = plt.cm.tab10(
    np.linspace(0, 1, 8)
)

# Plot each model's macro-average ROC curve
for index, (model_name, result) in enumerate(
    roc_results.items()
):

    ax.plot(
        result["fpr"],
        result["tpr"],
        color=colors[index],
        linewidth=2.4,
        label=model_name
    )

# Add random-classifier reference line
ax.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    color="gray",
    linewidth=1.5,
    label="Random classifier"
)

# Axis labels
ax.set_xlabel(
    "False Positive Rate",
    fontsize=13,
    fontweight="bold"
)

ax.set_ylabel(
    "True Positive Rate",
    fontsize=13,
    fontweight="bold"
)

# Figure title
ax.set_title(
    "Comparative ROC Curves of Eight Model Configurations",
    fontsize=15,
    fontweight="bold",
    pad=15
)

# Axis limits
ax.set_xlim(0, 1)
ax.set_ylim(0, 1.02)

# Axis tick formatting
ax.tick_params(
    axis="both",
    labelsize=11
)

# Subtle grid
ax.grid(
    True,
    linestyle=":",
    linewidth=0.7,
    alpha=0.45
)

# Legend outside the plotting area
ax.legend(
    loc="lower right",
    fontsize=9.5,
    frameon=True,
    framealpha=0.95,
    edgecolor="gray"
)

# Keep the ROC plot square
ax.set_aspect("equal", adjustable="box")

plt.tight_layout()

# ------------------------------------------------
# STEP 7: SAVE FIGURE IN THREE FORMATS
# ------------------------------------------------

figure_name = "Figure3_Eight_Model_Comparative_ROC"

png_path = os.path.join(
    GRAPH_DIR,
    figure_name + ".png"
)

tiff_path = os.path.join(
    GRAPH_DIR,
    figure_name + ".tiff"
)

pdf_path = os.path.join(
    GRAPH_DIR,
    figure_name + ".pdf"
)

# High-resolution publication exports
fig.savefig(
    png_path,
    dpi=600,
    bbox_inches="tight"
)

fig.savefig(
    tiff_path,
    dpi=600,
    bbox_inches="tight"
)

fig.savefig(
    pdf_path,
    bbox_inches="tight"
)

plt.show()

plt.close(fig)

# ------------------------------------------------
# STEP 8: SAVE NUMERICAL ROC-AUC RESULTS SEPARATELY
# ------------------------------------------------

# AUC values are saved in a table, not displayed on the figure
auc_table = pd.DataFrame([
    {
        "Model": model_name,
        "Macro_ROC_AUC": result["auc"]
    }
    for model_name, result in roc_results.items()
])

auc_csv_path = os.path.join(
    ROC_DIR,
    "Eight_Model_Macro_ROC_AUC.csv"
)

auc_table.to_csv(
    auc_csv_path,
    index=False
)

print("\n" + "=" * 85)
print("FIGURE 3 GENERATED SUCCESSFULLY")
print("=" * 85)

print("\nSaved files:")
print(png_path)
print(tiff_path)
print(pdf_path)
print(auc_csv_path)

print("\nROC-AUC results:")
print(auc_table.to_string(index=False))
```

```python

# ================================================================
# FIGURE 3: COMPARATIVE ROC CURVES OF ALL EIGHT MODEL CONFIGURATIONS
#
# Dataset       : Mendeley SEM - 237 original images
# Evaluation    : Stratified five-fold cross-validation
# ROC analysis  : One-vs-rest macro-average using pooled OOF probabilities
#
# IMPORTANT:
# 1. No model retraining.
# 2. No Random Forest model.
# 3. The diagonal line represents a random classifier (chance baseline).
# 4. No numerical AUC values or percentages displayed in the figure.
# 5. Numerical AUC values are saved separately in CSV format.
# ================================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import label_binarize
from sklearn.metrics import roc_curve, auc


# ------------------------------------------------
# STEP 1: DEFINE DIRECTORIES
# ------------------------------------------------

BASE_DIR = "/content/drive/MyDrive/Microplastics_SEM"

RESULTS_DIR = os.path.join(
    BASE_DIR, "AMAFFNet_Results"
)

VIS_DIR = os.path.join(
    BASE_DIR, "AMAFFNet_Visualization"
)

ROC_DIR = os.path.join(
    VIS_DIR, "ROC_Curves"
)

GRAPH_DIR = os.path.join(
    VIS_DIR, "Comparison_Graphs"
)

# Create output directories if they do not exist
os.makedirs(ROC_DIR, exist_ok=True)
os.makedirs(GRAPH_DIR, exist_ok=True)


# ------------------------------------------------
# STEP 2: DEFINE CLASS ORDER AND EXPECTED COUNTS
# ------------------------------------------------

CLASS_NAMES = ["Bead", "Fiber", "Fragment"]

EXPECTED_COUNTS = {
    "Bead": 51,
    "Fiber": 90,
    "Fragment": 96
}

EXPECTED_ROWS = 237


# ------------------------------------------------
# STEP 3: DEFINE EIGHT VERIFIED PREDICTION FILES
# ------------------------------------------------

MODEL_FILES = {

    "Xception (without SE)": {
        "path": os.path.join(
            RESULTS_DIR,
            "Xception",
            "Xception_All_Fold_Predictions.csv"
        ),
        "actual": "True_Class",
        "probabilities": [
            "Probability_Bead",
            "Probability_Fiber",
            "Probability_Fragment"
        ]
    },

    "Xception (with SE)": {
        "path": os.path.join(
            VIS_DIR,
            "Xception_SE_SE_Standardized_Predictions.csv"
        ),
        "actual": "Actual",
        "probabilities": [
            "Prob_Bead",
            "Prob_Fiber",
            "Prob_Fragment"
        ]
    },

    "DenseNet201 (without SE)": {
        "path": os.path.join(
            RESULTS_DIR,
            "DenseNet201_NoSE",
            "DenseNet201_NoSE_All_Fold_Predictions.csv"
        ),
        "actual": "Actual_Class",
        "probabilities": [
            "Bead_Probability",
            "Fiber_Probability",
            "Fragment_Probability"
        ]
    },

    "DenseNet201 (with SE)": {
        "path": os.path.join(
            VIS_DIR,
            "DenseNet201_SE_SE_Standardized_Predictions.csv"
        ),
        "actual": "Actual",
        "probabilities": [
            "Prob_Bead",
            "Prob_Fiber",
            "Prob_Fragment"
        ]
    },

    "MobileNetV2 (without SE)": {
        "path": os.path.join(
            RESULTS_DIR,
            "MobileNetV2_NoSE",
            "MobileNetV2_NoSE_All_Fold_Predictions.csv"
        ),
        "actual": "True_Class",
        "probabilities": [
            "Probability_Bead",
            "Probability_Fiber",
            "Probability_Fragment"
        ]
    },

    "MobileNetV2 (with SE)": {
        "path": os.path.join(
            VIS_DIR,
            "MobileNetV2_SE_SE_Standardized_Predictions.csv"
        ),
        "actual": "Actual",
        "probabilities": [
            "Prob_Bead",
            "Prob_Fiber",
            "Prob_Fragment"
        ]
    },

    "AMAFFNet (without SE)": {
        "path": os.path.join(
            ROC_DIR,
            "AMAFFNet_NoSE_All_Fold_Predictions.csv"
        ),
        "actual": "True_Class",
        "probabilities": [
            "Probability_Bead",
            "Probability_Fiber",
            "Probability_Fragment"
        ]
    },

    "AMAFFNet (with SE)": {
        "path": os.path.join(
            ROC_DIR,
            "AMAFFNet_SE_All_Fold_Predictions.csv"
        ),
        "actual": "True_Class",
        "probabilities": [
            "Probability_Bead",
            "Probability_Fiber",
            "Probability_Fragment"
        ]
    }
}


# ------------------------------------------------
# STEP 4: LOAD AND VALIDATE ALL PREDICTION FILES
# ------------------------------------------------

model_data = {}

print("=" * 85)
print("VALIDATING EIGHT MODEL PREDICTION FILES")
print("=" * 85)

for model_name, config in MODEL_FILES.items():

    file_path = config["path"]

    print("\nChecking:", model_name)

    # Verify that the prediction file exists
    if not os.path.isfile(file_path):
        raise FileNotFoundError(
            f"Prediction file not found:\n{file_path}"
        )

    # Read the pooled out-of-fold predictions
    df = pd.read_csv(file_path)

    # Check required columns
    required_columns = (
        [config["actual"]] + config["probabilities"]
    )

    missing_columns = [
        col for col in required_columns
        if col not in df.columns
    ]

    if missing_columns:
        raise ValueError(
            f"{model_name}: Missing columns {missing_columns}\n"
            f"Available columns: {df.columns.tolist()}"
        )

    # Extract actual class labels
    y_true = df[config["actual"]].copy()

    # Convert numeric labels into class names if necessary
    if pd.api.types.is_numeric_dtype(y_true):

        label_mapping = {
            0: "Bead",
            1: "Fiber",
            2: "Fragment"
        }

        y_true = y_true.map(label_mapping)

    y_true = y_true.astype(str).str.strip()

    # Extract predicted class probabilities
    y_prob = df[
        config["probabilities"]
    ].to_numpy(dtype=float)

    # Check number of predictions
    if len(df) != EXPECTED_ROWS:
        raise ValueError(
            f"{model_name}: Expected {EXPECTED_ROWS} rows, "
            f"found {len(df)}"
        )

    # Check for missing or unexpected class labels
    if y_true.isna().any():
        raise ValueError(
            f"{model_name}: Missing actual class labels."
        )

    if not set(y_true.unique()).issubset(set(CLASS_NAMES)):
        raise ValueError(
            f"{model_name}: Unexpected class labels "
            f"{y_true.unique().tolist()}"
        )

    # Check class distribution
    actual_counts = {
        class_name: int((y_true == class_name).sum())
        for class_name in CLASS_NAMES
    }

    if actual_counts != EXPECTED_COUNTS:
        raise ValueError(
            f"{model_name}: Incorrect class distribution "
            f"{actual_counts}"
        )

    # Check probability values
    if not np.isfinite(y_prob).all():
        raise ValueError(
            f"{model_name}: Non-finite probability values."
        )

    if np.any(y_prob < 0) or np.any(y_prob > 1):
        raise ValueError(
            f"{model_name}: Probabilities outside [0, 1]."
        )

    # Verify that each probability row sums approximately to one
    if not np.allclose(
        y_prob.sum(axis=1),
        1.0,
        atol=1e-4
    ):
        raise ValueError(
            f"{model_name}: Probability sums are invalid."
        )

    # Convert actual class labels to one-vs-rest binary format
    y_binary = label_binarize(
        y_true,
        classes=CLASS_NAMES
    )

    # Store validated data
    model_data[model_name] = {
        "y_binary": y_binary,
        "y_prob": y_prob
    }

    print("Rows:", len(df))
    print("Class distribution:", actual_counts)
    print("Probability validation: PASSED")

print("\nALL EIGHT MODELS PASSED VALIDATION.")


# ------------------------------------------------
# STEP 5: COMPUTE MACRO-AVERAGE ROC CURVES
# ------------------------------------------------

# Common false-positive-rate points for interpolation
mean_fpr = np.linspace(0, 1, 1000)

roc_results = {}

for model_name, data in model_data.items():

    y_binary = data["y_binary"]
    y_prob = data["y_prob"]

    class_tprs = []

    # Calculate one-vs-rest ROC for each class
    for class_index in range(len(CLASS_NAMES)):

        fpr, tpr, thresholds = roc_curve(
            y_binary[:, class_index],
            y_prob[:, class_index]
        )

        # Interpolate class ROC onto the common FPR grid
        interpolated_tpr = np.interp(
            mean_fpr,
            fpr,
            tpr
        )

        interpolated_tpr[0] = 0.0

        class_tprs.append(interpolated_tpr)

    # Calculate the macro-average ROC curve
    macro_tpr = np.mean(
        class_tprs,
        axis=0
    )

    macro_tpr[0] = 0.0
    macro_tpr[-1] = 1.0

    # Calculate macro-average AUC
    macro_auc = auc(
        mean_fpr,
        macro_tpr
    )

    roc_results[model_name] = {
        "fpr": mean_fpr,
        "tpr": macro_tpr,
        "auc": macro_auc
    }

    print(
        f"{model_name}: Macro ROC-AUC = {macro_auc:.6f}"
    )


# ------------------------------------------------
# STEP 6: PLOT ALL EIGHT ROC CURVES
# ------------------------------------------------

# Create a publication-quality figure
fig, ax = plt.subplots(
    figsize=(11, 8),
    dpi=150
)

# Eight distinct colours for the eight configurations
colors = plt.cm.tab10(
    np.linspace(0, 1, 8)
)

# Plot each model's macro-average ROC curve
for index, (model_name, result) in enumerate(
    roc_results.items()
):

    ax.plot(
        result["fpr"],
        result["tpr"],
        color=colors[index],
        linewidth=2.4,
        label=model_name,
        zorder=3
    )

# Add the chance-level reference line
# IMPORTANT: This is a random classifier, NOT Random Forest.
ax.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    color="black",
    linewidth=1.4,
    label="Chance-level classifier",
    zorder=2
)

# Set axis labels
ax.set_xlabel(
    "False Positive Rate",
    fontsize=13,
    fontweight="bold"
)

ax.set_ylabel(
    "True Positive Rate",
    fontsize=13,
    fontweight="bold"
)

# Set figure title
ax.set_title(
    "Comparative ROC Curves of Eight Model Configurations",
    fontsize=15,
    fontweight="bold",
    pad=15
)

# Set ROC axis limits
ax.set_xlim(0.0, 1.0)
ax.set_ylim(0.0, 1.02)

# Use consistent axis tick formatting
ax.tick_params(
    axis="both",
    labelsize=11,
    direction="out"
)

# Add subtle grid lines
ax.grid(
    True,
    linestyle=":",
    linewidth=0.7,
    alpha=0.4
)

# Display legend inside the lower-right region
ax.legend(
    loc="lower right",
    fontsize=9,
    frameon=True,
    framealpha=0.95,
    edgecolor="gray",
    ncol=1
)

# Maintain equal scaling for both ROC axes
ax.set_aspect("equal", adjustable="box")

# Improve spacing
plt.tight_layout()


# ------------------------------------------------
# STEP 7: SAVE FIGURE IN THREE FORMATS
# ------------------------------------------------

figure_name = "Figure3_Eight_Model_Comparative_ROC"

png_path = os.path.join(
    GRAPH_DIR,
    figure_name + ".png"
)

tiff_path = os.path.join(
    GRAPH_DIR,
    figure_name + ".tiff"
)

pdf_path = os.path.join(
    GRAPH_DIR,
    figure_name + ".pdf"
)

# Save high-resolution PNG
fig.savefig(
    png_path,
    dpi=600,
    bbox_inches="tight"
)

# Save high-resolution TIFF
fig.savefig(
    tiff_path,
    dpi=600,
    bbox_inches="tight"
)

# Save vector PDF
fig.savefig(
    pdf_path,
    bbox_inches="tight"
)

plt.show()

plt.close(fig)


# ------------------------------------------------
# STEP 8: SAVE NUMERICAL AUC RESULTS SEPARATELY
# ------------------------------------------------

# Numerical AUC values are stored in a CSV table
# They are deliberately not displayed on the ROC figure.

auc_table = pd.DataFrame([
    {
        "Model": model_name,
        "Macro_ROC_AUC": result["auc"]
    }
    for model_name, result in roc_results.items()
])

auc_csv_path = os.path.join(
    ROC_DIR,
    "Eight_Model_Macro_ROC_AUC.csv"
)

auc_table.to_csv(
    auc_csv_path,
    index=False
)


# ------------------------------------------------
# STEP 9: FINAL OUTPUT
# ------------------------------------------------

print("\n" + "=" * 85)
print("FIGURE 3 GENERATED SUCCESSFULLY")
print("=" * 85)

print("\nSaved files:")
print(png_path)
print(tiff_path)
print(pdf_path)
print(auc_csv_path)

print("\nROC-AUC results:")
print(
    auc_table.to_string(index=False)
)

print("\nAll eight model configurations were plotted.")
print("No Random Forest model was included.")
print("The diagonal line represents chance-level classification.")
```

```python

# ================================================================
# AMAFFNET_SE: RECURSIVE CONVOLUTIONAL LAYER INSPECTION
# Fold 1 | No retraining | No checkpoint reloading
#
# Purpose:
# 1. Search nested Functional models recursively.
# 2. Identify convolutional layers in all three SE backbones.
# 3. Display their names and output shapes.
# 4. Identify candidate final convolutional feature maps.
#
# Uses the models already loaded in the previous cell.
# ================================================================

import tensorflow as tf

# ------------------------------------------------
# 1. Confirm previously loaded models are available
# ------------------------------------------------

required_models = [
    "fusion_model",
    "xception_model",
    "densenet_model",
    "mobilenet_model"
]

for variable_name in required_models:
    if variable_name not in globals():
        raise NameError(
            f"{variable_name} is not available. "
            "Please run the model-loading cell first."
        )

backbone_models = {
    "Xception_SE": xception_model,
    "DenseNet201_SE": densenet_model,
    "MobileNetV2_SE": mobilenet_model
}

# ------------------------------------------------
# 2. Recursively inspect nested model layers
# ------------------------------------------------

def inspect_layers_recursively(model, prefix=""):

    results = []

    for layer in model.layers:

        layer_path = (
            f"{prefix}/{layer.name}"
            if prefix else layer.name
        )

        # Record convolutional layers.
        if isinstance(
            layer,
            (
                tf.keras.layers.Conv2D,
                tf.keras.layers.DepthwiseConv2D,
                tf.keras.layers.SeparableConv2D
            )
        ):
            results.append({
                "path": layer_path,
                "name": layer.name,
                "type": layer.__class__.__name__,
                "layer": layer
            })

        # Search inside nested Functional or Sequential models.
        if isinstance(
            layer,
            (tf.keras.Model, tf.keras.Sequential)
        ):
            results.extend(
                inspect_layers_recursively(
                    layer,
                    prefix=layer_path
                )
            )

    return results

# ------------------------------------------------
# 3. Inspect all three SE backbones
# ------------------------------------------------

print("=" * 80)
print("RECURSIVE CONVOLUTIONAL LAYER INSPECTION")
print("=" * 80)

final_conv_layers = {}

for model_name, backbone in backbone_models.items():

    print("\n" + "=" * 80)
    print("MODEL:", model_name)
    print("=" * 80)

    conv_layers = inspect_layers_recursively(backbone)

    if not conv_layers:
        print("No convolutional layers found.")
        continue

    for index, item in enumerate(conv_layers, start=1):

        layer = item["layer"]

        print(
            f"{index:03d}. "
            f"{item['path']} | "
            f"{item['type']}"
        )

    # The last convolutional layer in recursive model order
    # is a candidate. We will verify graph connectivity next.
    final_item = conv_layers[-1]

    final_conv_layers[model_name] = final_item

    print("\nCandidate final convolutional layer:")
    print("Name:", final_item["name"])
    print("Full path:", final_item["path"])

# ------------------------------------------------
# 4. Verify candidate layer output shapes
# ------------------------------------------------

print("\n" + "=" * 80)
print("CANDIDATE FINAL CONVOLUTIONAL LAYERS")
print("=" * 80)

for model_name, item in final_conv_layers.items():

    print("\n", model_name)
    print("Layer:", item["name"])

    try:
        print("Output shape:", item["layer"].output.shape)
    except Exception as error:
        print("Output shape unavailable:", str(error))

# ------------------------------------------------
# 5. Verify SE-related layers in each backbone
# ------------------------------------------------

print("\n" + "=" * 80)
print("SE LAYER VERIFICATION")
print("=" * 80)

for model_name, backbone in backbone_models.items():

    all_layers = inspect_layers_recursively(backbone)

    se_layers = [
        item for item in all_layers
        if any(
            term in item["name"].lower()
            for term in ["se_", "squeeze", "excite"]
        )
    ]

    print("\n", model_name)

    if se_layers:
        for item in se_layers:
            print(item["path"])
    else:
        print(
            "No explicitly named SE convolutional layers found. "
            "This does not alone establish absence of SE blocks."
        )

print("\n" + "=" * 80)
print("INSPECTION COMPLETED")
print("No retraining performed.")
print("No weights modified.")
print("=" * 80)
```

```python

# ================================================================
# AMAFFNET_SE: GRADIENT CONNECTIVITY DIAGNOSTIC
# Reuses the already loaded models and recursive layer inspection.
# No retraining and no weight updates.
# ================================================================

import tensorflow as tf
from tensorflow.keras.models import Model

# ------------------------------------------------
# 1. Retrieve the candidate convolutional layers
# ------------------------------------------------

candidate_layers = {
    "Xception_SE": "block14_sepconv2",
    "DenseNet201_SE": "conv5_block32_2_conv",
    "MobileNetV2_SE": "Conv_1"
}

backbone_models = {
    "Xception_SE": xception_model,
    "DenseNet201_SE": densenet_model,
    "MobileNetV2_SE": mobilenet_model
}

# ------------------------------------------------
# 2. Find the nested layer by its full model path
# ------------------------------------------------

def find_nested_layer(model, target_name):

    for layer in model.layers:

        if layer.name == target_name:
            return layer

        if isinstance(layer, tf.keras.Model):
            found = find_nested_layer(layer, target_name)

            if found is not None:
                return found

    return None

# ------------------------------------------------
# 3. Display the final convolutional layer candidates
# ------------------------------------------------

print("=" * 75)
print("CHECKING FINAL CONVOLUTIONAL LAYER REFERENCES")
print("=" * 75)

for name, backbone in backbone_models.items():

    target_name = candidate_layers[name]
    layer = find_nested_layer(backbone, target_name)

    if layer is None:
        raise ValueError(
            f"Could not locate {target_name} in {name}"
        )

    print(
        f"{name}: {layer.name} | "
        f"Output shape: {layer.output.shape}"
    )

# ------------------------------------------------
# 4. Test the existing feature extractors and fusion model
# ------------------------------------------------

print("\n" + "=" * 75)
print("CHECKING FEATURE EXTRACTOR AND FUSION CONNECTION")
print("=" * 75)

# Reuse the existing feature extractors created previously.
feature_models = {
    "Xception_SE": Model(
        xception_model.input,
        xception_model.get_layer("Feature_Vector").output
    ),
    "DenseNet201_SE": Model(
        densenet_model.input,
        densenet_model.get_layer("Feature_Vector").output
    ),
    "MobileNetV2_SE": Model(
        mobilenet_model.input,
        mobilenet_model.get_layer("Feature_Vector").output
    )
}

# ------------------------------------------------
# 5. Build an end-to-end prediction using image inputs
# ------------------------------------------------

image_inputs = [
    tf.keras.Input(
        shape=backbone.input_shape[1:],
        name=f"{name}_Image"
    )
    for name, backbone in backbone_models.items()
]

feature_outputs = [
    feature_models[name](image_input)
    for name, image_input in zip(
        backbone_models.keys(),
        image_inputs
    )
]

final_output = fusion_model(feature_outputs)

end_to_end_model = Model(
    inputs=image_inputs,
    outputs=final_output
)

print("End-to-end output:", end_to_end_model.output_shape)

# ------------------------------------------------
# 6. Test gradients with respect to each image input
# ------------------------------------------------

dummy_images = [
    tf.zeros(
        (1,) + tuple(backbone.input_shape[1:]),
        dtype=tf.float32
    )
    for backbone in backbone_models.values()
]

with tf.GradientTape(persistent=True) as tape:

    for image in dummy_images:
        tape.watch(image)

    predictions = end_to_end_model(
        dummy_images,
        training=False
    )

    target_class = tf.argmax(
        predictions[0],
        output_type=tf.int32
    )

    target_score = predictions[0, target_class]

print("\n" + "=" * 75)
print("IMAGE-TO-FUSION GRADIENT TEST")
print("=" * 75)

for name, image in zip(backbone_models.keys(), dummy_images):

    gradient = tape.gradient(target_score, image)

    if gradient is None:
        print(name, ": DISCONNECTED")
    else:
        print(
            name,
            ": CONNECTED | Gradient shape:",
            gradient.shape
        )

del tape

print("\n" + "=" * 75)
print("DIAGNOSTIC COMPLETED")
print("No retraining performed.")
print("=" * 75)
```

```python

# ================================================================
# AMAFFNET_SE: BRANCH-SPECIFIC GRAD-CAM
# Uses the FINAL FUSED CLASS SCORE for all three branches
# Uses held-out Fold 1 images
# No retraining and no weight modification
# ================================================================

import os
import numpy as np
import pandas as pd
import tensorflow as tf
import matplotlib.pyplot as plt

from PIL import Image
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input
from tensorflow.keras.applications.xception import preprocess_input as xception_preprocess
from tensorflow.keras.applications.densenet import preprocess_input as densenet_preprocess
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input as mobilenet_preprocess

# ------------------------------------------------
# 1. CONFIGURATION
# ------------------------------------------------

BASE_DIR = "/content/drive/MyDrive/Microplastics_SEM"

TEST_CSV = os.path.join(
    BASE_DIR, "AMAFFNet_5Fold_Splits", "fold_1_test.csv"
)

OUTPUT_DIR = os.path.join(
    BASE_DIR, "AMAFFNet_Visualization", "GradCAM_AMAFFNet_SE"
)
os.makedirs(OUTPUT_DIR, exist_ok=True)

CLASS_NAMES = ["Bead", "Fiber", "Fragment"]
IMG_SIZES = {
    "Xception": (299, 299),
    "DenseNet201": (224, 224),
    "MobileNetV2": (160, 160)
}

PREPROCESS = {
    "Xception": xception_preprocess,
    "DenseNet201": densenet_preprocess,
    "MobileNetV2": mobilenet_preprocess
}

# Models must already be loaded in the notebook.
required_models = [
    "xception_model", "densenet_model",
    "mobilenet_model", "fusion_model"
]
for model_name in required_models:
    if model_name not in globals():
        raise NameError(
            f"{model_name} is not loaded. Load the existing models first."
        )

# ------------------------------------------------
# 2. BUILD BRANCH MODELS
# Each branch returns:
#   (256-dimensional feature vector, Grad-CAM activation)
# The same backbone invocation produces both outputs.
# ------------------------------------------------

def build_xception_branch(backbone):
    image = Input(shape=(299, 299, 3), name="Xception_GradCAM_Input")

    base = backbone.get_layer("xception")
    conv_map = base(image, training=False)

    x = backbone.get_layer("global_average_pooling2d")(conv_map)
    x = backbone.get_layer("batch_normalization_4")(x, training=False)
    x = backbone.get_layer("dense")(x)
    x = backbone.get_layer("dropout")(x, training=False)
    feature = backbone.get_layer("Feature_Vector")(x)

    return Model(image, [feature, conv_map], name="Xception_GradCAM_Branch")


def build_se_branch(backbone, nested_name, input_size, branch_name):
    image = Input(
        shape=(input_size, input_size, 3),
        name=f"{branch_name}_GradCAM_Input"
    )

    # Obtain the convolutional feature map from the nested CNN.
    base = backbone.get_layer(nested_name)
    base_map = base(image, training=False)

    # Reconstruct the trained SE attention pathway using the
    # original trained layers and weights.
    x = backbone.get_layer("global_average_pooling2d")(base_map)
    x = backbone.get_layer("dense")(x)
    x = backbone.get_layer("dense_1")(x)
    attention = backbone.get_layer("reshape")(x)

    # Post-SE feature map is the Grad-CAM target.
    attended_map = backbone.get_layer("multiply")(
        [base_map, attention]
    )

    x = backbone.get_layer("global_average_pooling2d_1")(attended_map)
    x = backbone.get_layer("batch_normalization")(x, training=False)
    x = backbone.get_layer("dense_2")(x)
    x = backbone.get_layer("dropout")(x, training=False)
    feature = backbone.get_layer("Feature_Vector")(x)

    return Model(
        image, [feature, attended_map],
        name=f"{branch_name}_GradCAM_Branch"
    )


# Build the three branches with their existing trained weights.
branch_models = {
    "Xception": build_xception_branch(xception_model),

    "DenseNet201": build_se_branch(
        densenet_model, "densenet201", 224, "DenseNet201"
    ),

    "MobileNetV2": build_se_branch(
        mobilenet_model, "mobilenetv2_1.00_224", 160, "MobileNetV2"
    )
}

# ------------------------------------------------
# 3. LOAD FOLD 1 HELD-OUT IMAGES
# ------------------------------------------------

if not os.path.exists(TEST_CSV):
    raise FileNotFoundError(f"Fold 1 test CSV not found: {TEST_CSV}")

test_df = pd.read_csv(TEST_CSV)

if "Image_Path" not in test_df.columns or "Class" not in test_df.columns:
    raise ValueError(
        "The Fold 1 test CSV must contain Image_Path and Class columns."
    )

test_df = test_df.dropna(subset=["Image_Path", "Class"]).copy()
test_df["Class"] = test_df["Class"].astype(str)

def load_preprocessed_image(image_path, branch):
    size = IMG_SIZES[branch]

    # Load original image and resize using the backbone input size.
    image = Image.open(image_path).convert("RGB").resize(
        (size[1], size[0])
    )
    array = np.asarray(image, dtype=np.float32)

    # Apply the exact Keras application preprocessing function.
    array = PREPROCESS[branch](array)
    return np.expand_dims(array, axis=0), np.asarray(image)


# ------------------------------------------------
# 4. VERIFY FUSED MODEL OUTPUT AND SELECT
#    ONE CORRECTLY CLASSIFIED IMAGE PER CLASS
# ------------------------------------------------

selected = {}

print("\nSelecting correctly classified Fold 1 images...")

for _, row in test_df.iterrows():

    true_class = row["Class"]
    image_path = str(row["Image_Path"])

    if true_class not in CLASS_NAMES or true_class in selected:
        continue

    if not os.path.exists(image_path):
        continue

    try:
        branch_features = []

        for branch_name, branch_model in branch_models.items():
            batch, _ = load_preprocessed_image(image_path, branch_name)
            feature, _ = branch_model(batch, training=False)
            branch_features.append(feature)

        # Final AMAFFNet_SE fused prediction.
        probabilities = fusion_model(
            branch_features, training=False
        ).numpy()[0]

        predicted_index = int(np.argmax(probabilities))
        predicted_class = CLASS_NAMES[predicted_index]

        # Select only correctly classified held-out examples.
        if predicted_class == true_class:
            selected[true_class] = {
                "path": image_path,
                "probabilities": probabilities,
                "predicted_class": predicted_class
            }

    except Exception as error:
        print("Image skipped:", image_path, "|", str(error))

    if len(selected) == len(CLASS_NAMES):
        break

if len(selected) != len(CLASS_NAMES):
    raise RuntimeError(
        "Could not find correctly classified held-out examples "
        "for all three classes. Selected: " + str(list(selected.keys()))
    )

for class_name, item in selected.items():
    print(
        class_name,
        "| Prediction:", item["predicted_class"],
        "| Probabilities:", np.round(item["probabilities"], 4)
    )

# ------------------------------------------------
# 5. GRAD-CAM FUNCTION
# Gradients are calculated from the fused class score.
# ------------------------------------------------

def generate_gradcam(branch_name, image_batch, target_class_index):

    branch_model = branch_models[branch_name]

    with tf.GradientTape() as tape:

        # Extract feature vector and convolutional activation.
        feature_vector, conv_activation = branch_model(
            image_batch, training=False
        )

        # Watch the actual intermediate activation.
        tape.watch(conv_activation)

        # Use the final fused AMAFFNet_SE prediction.
        all_features = []

        for name in ["Xception", "DenseNet201", "MobileNetV2"]:
            if name == branch_name:
                all_features.append(feature_vector)
            else:
                # Other branches provide their normal feature vectors.
                other_batch, _ = load_preprocessed_image(
                    CURRENT_IMAGE_PATH, name
                )
                other_feature, _ = branch_models[name](
                    other_batch, training=False
                )
                all_features.append(other_feature)

        fused_probabilities = fusion_model(
            all_features, training=False
        )

        class_score = fused_probabilities[:, target_class_index]

    # Calculate gradients of the fused class score
    # with respect to this branch's convolutional activation.
    gradients = tape.gradient(class_score, conv_activation)

    if gradients is None:
        raise RuntimeError(
            f"Disconnected Grad-CAM gradients for {branch_name}."
        )

    # Global-average-pool the gradients.
    weights = tf.reduce_mean(
        gradients, axis=(1, 2), keepdims=True
    )

    # Weighted sum of feature maps.
    cam = tf.reduce_sum(
        weights * conv_activation, axis=-1
    )

    # ReLU and normalize the heatmap.
    cam = tf.nn.relu(cam)[0]
    cam = cam / (tf.reduce_max(cam) + tf.keras.backend.epsilon())

    return cam.numpy()


# ------------------------------------------------
# 6. GENERATE THREE-BRANCH GRAD-CAM FOR EACH CLASS
# ------------------------------------------------

import cv2

all_results = {}

for true_class in CLASS_NAMES:

    print("\nGenerating Grad-CAM:", true_class)

    CURRENT_IMAGE_PATH = selected[true_class]["path"]
    target_index = CLASS_NAMES.index(true_class)

    original_image = np.asarray(
        Image.open(CURRENT_IMAGE_PATH).convert("RGB")
    )

    branch_heatmaps = {}
    branch_overlays = {}

    for branch_name in ["Xception", "DenseNet201", "MobileNetV2"]:

        image_batch, _ = load_preprocessed_image(
            CURRENT_IMAGE_PATH, branch_name
        )

        heatmap = generate_gradcam(
            branch_name, image_batch, target_index
        )

        # Resize the heatmap to the original image dimensions.
        heatmap = cv2.resize(
            heatmap,
            (original_image.shape[1], original_image.shape[0])
        )

        heatmap = np.clip(heatmap, 0, 1)
        branch_heatmaps[branch_name] = heatmap

        # Create a color heatmap and overlay.
        heatmap_uint8 = np.uint8(255 * heatmap)
        colored_map = cv2.applyColorMap(
            heatmap_uint8, cv2.COLORMAP_JET
        )
        colored_map = cv2.cvtColor(
            colored_map, cv2.COLOR_BGR2RGB
        )

        overlay = cv2.addWeighted(
            original_image, 0.55, colored_map, 0.45, 0
        )

        branch_overlays[branch_name] = overlay

        # Save individual heatmap and overlay.
        safe_class = true_class.replace(" ", "_")
        cv2.imwrite(
            os.path.join(
                OUTPUT_DIR,
                f"{safe_class}_{branch_name}_Heatmap.png"
            ),
            heatmap_uint8
        )

        Image.fromarray(overlay).save(
            os.path.join(
                OUTPUT_DIR,
                f"{safe_class}_{branch_name}_Overlay.png"
            )
        )

    all_results[true_class] = {
        "original": original_image,
        "heatmaps": branch_heatmaps,
        "overlays": branch_overlays
    }

# ------------------------------------------------
# 7. CREATE COMBINED GRAD-CAM FIGURE
# Rows: Bead, Fiber, Fragment
# Columns: Original, Xception, DenseNet201, MobileNetV2
# ------------------------------------------------

fig, axes = plt.subplots(
    3, 4, figsize=(16, 11)
)

column_titles = [
    "Original SEM Image",
    "Xception",
    "DenseNet201",
    "MobileNetV2"
]

for col, title in enumerate(column_titles):
    axes[0, col].set_title(title, fontsize=13, fontweight="bold")

for row, class_name in enumerate(CLASS_NAMES):

    result = all_results[class_name]

    axes[row, 0].imshow(result["original"])
    axes[row, 0].set_ylabel(
        class_name, fontsize=13, fontweight="bold"
    )

    for col, branch_name in enumerate(
        ["Xception", "DenseNet201", "MobileNetV2"], start=1
    ):
        axes[row, col].imshow(
            result["overlays"][branch_name]
        )

    for col in range(4):
        axes[row, col].axis("off")

plt.suptitle(
    "Grad-CAM Visualization of AMAFFNet_SE",
    fontsize=16,
    fontweight="bold",
    y=0.98
)

plt.tight_layout(rect=[0, 0, 1, 0.96])

figure_path = os.path.join(
    OUTPUT_DIR, "Figure_GradCAM_AMAFFNet_SE.png"
)

plt.savefig(
    figure_path,
    dpi=600,
    bbox_inches="tight"
)
plt.show()

# Also save publication-quality TIFF and PDF.
fig.savefig(
    os.path.join(
        OUTPUT_DIR, "Figure_GradCAM_AMAFFNet_SE.tiff"
    ),
    dpi=600,
    bbox_inches="tight"
)

fig.savefig(
    os.path.join(
        OUTPUT_DIR, "Figure_GradCAM_AMAFFNet_SE.pdf"
    ),
    dpi=600,
    bbox_inches="tight"
)

# ------------------------------------------------
# 8. SAVE SELECTED IMAGE RECORDS
# ------------------------------------------------

selection_df = pd.DataFrame([
    {
        "True_Class": class_name,
        "Predicted_Class": selected[class_name]["predicted_class"],
        "Image_Path": selected[class_name]["path"],
        "Probability_Bead": selected[class_name]["probabilities"][0],
        "Probability_Fiber": selected[class_name]["probabilities"][1],
        "Probability_Fragment": selected[class_name]["probabilities"][2]
    }
    for class_name in CLASS_NAMES
])

selection_df.to_csv(
    os.path.join(OUTPUT_DIR, "GradCAM_Selected_Images.csv"),
    index=False
)

print("\n" + "=" * 75)
print("AMAFFNET_SE GRAD-CAM COMPLETED")
print("=" * 75)
print("Output directory:", OUTPUT_DIR)
print("Combined figure:", figure_path)
print("Saved formats: PNG, TIFF, PDF")
print("Individual branch heatmaps and overlays saved.")
print("No model retraining performed.")
```

```python

# =====================================================================
# AMAFFNET: COMPLETE COMPUTATIONAL COMPLEXITY ANALYSIS
# Eight configurations: 6 individual backbones + 2 AMAFFNet pipelines
#
# IMPORTANT:
# - Uses existing Fold-1 trained checkpoints only.
# - No retraining or weight modification.
# - AMAFFNet FLOPs and latency include all three feature extractors
#   followed by the fusion classifier.
# - Fusion-head parameters are NOT reported as complete AMAFFNet params.
# =====================================================================

import os
import gc
import time
import warnings
import numpy as np
import pandas as pd
import tensorflow as tf

from tensorflow.python.framework.convert_to_constants import (
    convert_variables_to_constants_v2
)

warnings.filterwarnings("ignore")

# ------------------------- 1. Configuration ----------------------------

BASE = "/content/drive/MyDrive/Microplastics_SEM"

OUTPUT_DIR = os.path.join(
    BASE, "Computational_Complexity_Results"
)
os.makedirs(OUTPUT_DIR, exist_ok=True)

# Exact verified fold-1 checkpoint paths.
BACKBONES = {
    "Xception_NoSE": {
        "path": f"{BASE}/AMAFFNet_Weights/Xception/best_xception_fold_1.keras",
        "size": (299, 299)
    },
    "Xception_SE": {
        "path": f"{BASE}/AMAFFNet_Weights/Xception_SE/best_xception_se_fold_1.keras",
        "size": (299, 299)
    },
    "DenseNet201_NoSE": {
        "path": f"{BASE}/AMAFFNet_Weights/DenseNet201_NoSE/best_densenet201_no_se_fold_1.keras",
        "size": (224, 224)
    },
    "DenseNet201_SE": {
        "path": f"{BASE}/AMAFFNet_Weights/DenseNet201/best_densenet201_fold_1.keras",
        "size": (224, 224)
    },
    "MobileNetV2_NoSE": {
        "path": f"{BASE}/AMAFFNet_Weights/MobileNetV2_NoSE/best_mobilenetv2_nose_fold_1.keras",
        "size": (160, 160)
    },
    "MobileNetV2_SE": {
        "path": f"{BASE}/AMAFFNet_Weights/MobileNetV2/best_mobilenetv2_fold_1.keras",
        "size": (160, 160)
    }
}

FUSION_MODELS = {
    "AMAFFNet_NoSE":
        f"{BASE}/AMAFFNet_Fusion_Results/AMAFFNet_NoSE/Fold_1/AMAFFNet_Final_Model.keras",

    "AMAFFNet_SE":
        f"{BASE}/AMAFFNet_Fusion_Results/AMAFFNet_SE/Fold_1/AMAFFNet_Final_Model.keras"
}

# Number of warm-up and timed inference iterations.
WARMUP_STEPS = 10
TIMED_STEPS = 50

# ------------------------- 2. Hardware details -------------------------

print("=" * 85)
print("COMPUTATIONAL COMPLEXITY ANALYSIS")
print("=" * 85)

print("TensorFlow version:", tf.__version__)
print("GPU devices:", tf.config.list_physical_devices("GPU"))

# ------------------------- 3. FLOPs calculation -------------------------

def calculate_flops(model, input_shapes):
    """
    Estimate floating-point operations using TensorFlow graph profiling.
    input_shapes is a list of complete input shapes including batch size.
    """

    try:
        specs = [
            tf.TensorSpec(shape=shape, dtype=tf.float32)
            for shape in input_shapes
        ]

        # Create a traced function for the model.
        if len(specs) == 1:
            traced = tf.function(
                lambda x: model(x, training=False)
            ).get_concrete_function(specs[0])
        else:
            traced = tf.function(
                lambda *args: model(list(args), training=False)
            ).get_concrete_function(*specs)

        # Freeze variables into constants for graph profiling.
        frozen = convert_variables_to_constants_v2(traced)

        graph = frozen.graph

        # Count floating-point operations.
        options = (
            tf.compat.v1.profiler.ProfileOptionBuilder
            .float_operation()
        )
        options["output"] = "none"

        profile = tf.compat.v1.profiler.profile(
            graph=graph,
            options=options
        )

        if profile is None:
            return None

        return int(profile.total_float_ops)

    except Exception as error:
        print("FLOPs calculation warning:", str(error)[:250])
        return None


# ------------------------- 4. Inference timing -------------------------

def measure_latency(function, inputs):
    """
    Measure average single-image latency after warm-up.
    .numpy() forces completion of the inference operation.
    """

    # Warm-up to reduce initialization effects.
    for _ in range(WARMUP_STEPS):
        output = function(*inputs) if len(inputs) > 1 else function(inputs[0])
        if isinstance(output, (list, tuple)):
            _ = [x.numpy() for x in output]
        else:
            _ = output.numpy()

    # Timed inference repetitions.
    timings = []

    for _ in range(TIMED_STEPS):

        start = time.perf_counter()

        output = function(*inputs) if len(inputs) > 1 else function(inputs[0])

        # Synchronize execution before recording elapsed time.
        if isinstance(output, (list, tuple)):
            _ = [x.numpy() for x in output]
        else:
            _ = output.numpy()

        timings.append((time.perf_counter() - start) * 1000)

    return float(np.mean(timings)), float(np.std(timings, ddof=1))


# ------------------------- 5. GPU memory helper ------------------------

def reset_gpu_memory():
    """Reset peak GPU memory statistics when a GPU is available."""

    gpus = tf.config.list_physical_devices("GPU")

    if gpus:
        try:
            tf.config.experimental.reset_memory_stats("GPU:0")
        except Exception:
            pass


def get_peak_gpu_memory_mb():
    """Return TensorFlow-tracked peak GPU memory in MB."""

    gpus = tf.config.list_physical_devices("GPU")

    if not gpus:
        return None

    try:
        info = tf.config.experimental.get_memory_info("GPU:0")
        return round(info["peak"] / (1024 ** 2), 3)
    except Exception:
        return None


# ------------------------- 6. Individual backbone analysis ------------

individual_results = []

print("\n" + "=" * 85)
print("INDIVIDUAL BACKBONE ANALYSIS")
print("=" * 85)

for model_name, config in BACKBONES.items():

    model_path = config["path"]
    height, width = config["size"]

    print("\nAnalyzing:", model_name)

    if not os.path.exists(model_path):
        print("CHECKPOINT MISSING:", model_path)
        continue

    model = None

    try:
        # Load existing trained classifier.
        model = tf.keras.models.load_model(
            model_path,
            compile=False,
            safe_mode=False
        )

        # Build a feature extractor ending at Feature_Vector.
        feature_layer = model.get_layer("Feature_Vector")

        extractor = tf.keras.Model(
            inputs=model.inputs,
            outputs=feature_layer.output,
            name=model_name + "_Feature_Extractor"
        )

        # Generate one synthetic input for benchmarking.
        # This is a performance benchmark, not a test-set evaluation.
        dummy_input = tf.random.uniform(
            (1, height, width, 3),
            dtype=tf.float32
        )

        # Count complete classifier model parameters.
        total_params = int(model.count_params())

        # Count feature extractor parameters used by AMAFFNet.
        feature_params = int(extractor.count_params())

        # FLOPs for the individual classifier.
        classifier_flops = calculate_flops(
            model,
            [(1, height, width, 3)]
        )

        # FLOPs for the feature extractor used by AMAFFNet.
        extractor_flops = calculate_flops(
            extractor,
            [(1, height, width, 3)]
        )

        # Benchmark the individual classifier.
        reset_gpu_memory()

        latency_mean, latency_sd = measure_latency(
            lambda x: model(x, training=False),
            [dummy_input]
        )

        peak_memory = get_peak_gpu_memory_mb()

        # Checkpoint storage size.
        file_size = os.path.getsize(model_path) / (1024 ** 2)

        individual_results.append({
            "Model": model_name,
            "Input_Size": f"{height}x{width}",
            "Classifier_Parameters": total_params,
            "Feature_Extractor_Parameters": feature_params,
            "Classifier_FLOPs": classifier_flops,
            "Feature_Extractor_FLOPs": extractor_flops,
            "Latency_Mean_ms": latency_mean,
            "Latency_SD_ms": latency_sd,
            "Peak_GPU_Memory_MB": peak_memory,
            "Checkpoint_Size_MB": file_size
        })

        print("Parameters:", f"{total_params:,}")
        print("Feature extractor parameters:", f"{feature_params:,}")
        print("Classifier FLOPs:", classifier_flops)
        print("Feature extractor FLOPs:", extractor_flops)
        print("Latency:", f"{latency_mean:.3f} ± {latency_sd:.3f} ms")
        print("Peak GPU memory (MB):", peak_memory)

    except Exception as error:
        print("ERROR:", repr(error))

    finally:
        del model
        if "extractor" in locals():
            del extractor
        gc.collect()
        tf.keras.backend.clear_session()


# ------------------------- 7. Complete AMAFFNet analysis ---------------

fusion_results = []

print("\n" + "=" * 85)
print("COMPLETE AMAFFNET PIPELINE ANALYSIS")
print("=" * 85)

# Exact branch-to-backbone mapping.
BRANCH_MAPPING = {
    "AMAFFNet_NoSE": [
        "Xception_NoSE",
        "DenseNet201_NoSE",
        "MobileNetV2_NoSE"
    ],
    "AMAFFNet_SE": [
        "Xception_SE",
        "DenseNet201_SE",
        "MobileNetV2_SE"
    ]
}

for fusion_name, fusion_path in FUSION_MODELS.items():

    print("\nAnalyzing:", fusion_name)

    if not os.path.exists(fusion_path):
        print("FUSION MODEL MISSING:", fusion_path)
        continue

    loaded_models = []
    extractors = []

    try:
        # Load the three corresponding trained backbone models.
        for backbone_name in BRANCH_MAPPING[fusion_name]:

            config = BACKBONES[backbone_name]

            backbone = tf.keras.models.load_model(
                config["path"],
                compile=False,
                safe_mode=False
            )

            # Verify 256-dimensional feature output.
            feature_extractor = tf.keras.Model(
                inputs=backbone.inputs,
                outputs=backbone.get_layer("Feature_Vector").output
            )

            if int(feature_extractor.output_shape[-1]) != 256:
                raise ValueError(
                    f"{backbone_name} does not produce 256 features."
                )

            loaded_models.append(backbone)
            extractors.append(feature_extractor)

        # Load the corresponding fusion classifier.
        fusion = tf.keras.models.load_model(
            fusion_path,
            compile=False,
            safe_mode=False
        )

        # Confirm all three fusion inputs are 256-dimensional.
        expected_shapes = fusion.input_shape

        if len(expected_shapes) != 3 or any(
            shape[-1] != 256 for shape in expected_shapes
        ):
            raise ValueError(
                f"Unexpected fusion input dimensions: {expected_shapes}"
            )

        # Count parameters actually used in the complete pipeline.
        # Use feature extractor parameters, not unused backbone
        # classification-head parameters.
        branch_params = sum(
            int(extractor.count_params())
            for extractor in extractors
        )

        fusion_params = int(fusion.count_params())

        complete_params = branch_params + fusion_params

        # Calculate FLOPs of each feature extraction branch.
        branch_flops = []

        for extractor, config in zip(
            extractors,
            [BACKBONES[name] for name in BRANCH_MAPPING[fusion_name]]
        ):

            h, w = config["size"]

            value = calculate_flops(
                extractor,
                [(1, h, w, 3)]
            )

            branch_flops.append(value)

        # Calculate fusion-head FLOPs.
        fusion_flops = calculate_flops(
            fusion,
            [(1, 256), (1, 256), (1, 256)]
        )

        # Sum branch FLOPs and fusion-head FLOPs.
        if all(value is not None for value in branch_flops) and fusion_flops is not None:
            complete_flops = sum(branch_flops) + fusion_flops
        else:
            complete_flops = None

        # Prepare synthetic inputs for the complete pipeline.
        pipeline_inputs = [
            tf.random.uniform(
                (1, BACKBONES[name]["size"][0],
                 BACKBONES[name]["size"][1], 3),
                dtype=tf.float32
            )
            for name in BRANCH_MAPPING[fusion_name]
        ]

        # Define complete end-to-end inference.
        def complete_pipeline(x1, x2, x3):
            f1 = extractors[0](x1, training=False)
            f2 = extractors[1](x2, training=False)
            f3 = extractors[2](x3, training=False)

            return fusion([f1, f2, f3], training=False)

        reset_gpu_memory()

        # Measure complete pipeline latency.
        latency_mean, latency_sd = measure_latency(
            complete_pipeline,
            pipeline_inputs
        )

        peak_memory = get_peak_gpu_memory_mb()

        # Total saved checkpoint storage:
        # three backbone checkpoints + one fusion checkpoint.
        checkpoint_paths = [
            BACKBONES[name]["path"]
            for name in BRANCH_MAPPING[fusion_name]
        ] + [fusion_path]

        total_storage_mb = sum(
            os.path.getsize(path) / (1024 ** 2)
            for path in checkpoint_paths
        )

        fusion_results.append({
            "Model": fusion_name,
            "Branch_Parameters": branch_params,
            "Fusion_Parameters": fusion_params,
            "Complete_Pipeline_Parameters": complete_params,
            "Branch_FLOPs": str(branch_flops),
            "Fusion_FLOPs": fusion_flops,
            "Complete_Pipeline_FLOPs": complete_flops,
            "Latency_Mean_ms": latency_mean,
            "Latency_SD_ms": latency_sd,
            "Peak_GPU_Memory_MB": peak_memory,
            "Total_Checkpoint_Storage_MB": total_storage_mb
        })

        print("Branch parameters:", f"{branch_params:,}")
        print("Fusion parameters:", f"{fusion_params:,}")
        print("Complete pipeline parameters:", f"{complete_params:,}")
        print("Branch FLOPs:", branch_flops)
        print("Fusion FLOPs:", fusion_flops)
        print("Complete pipeline FLOPs:", complete_flops)
        print("Latency:", f"{latency_mean:.3f} ± {latency_sd:.3f} ms")
        print("Peak GPU memory (MB):", peak_memory)
        print("Total checkpoint storage (MB):", round(total_storage_mb, 3))

    except Exception as error:
        print("AMAF FNET PIPELINE ERROR:", repr(error))

    finally:
        # Release loaded models before the next configuration.
        for item in loaded_models:
            del item

        for item in extractors:
            del item

        if "fusion" in locals():
            del fusion

        gc.collect()
        tf.keras.backend.clear_session()


# ------------------------- 8. Save reports ------------------------------

individual_df = pd.DataFrame(individual_results)
fusion_df = pd.DataFrame(fusion_results)

individual_csv = os.path.join(
    OUTPUT_DIR, "Individual_Backbone_Complexity.csv"
)

fusion_csv = os.path.join(
    OUTPUT_DIR, "Complete_AMAFFNet_Pipeline_Complexity.csv"
)

individual_df.to_csv(individual_csv, index=False)
fusion_df.to_csv(fusion_csv, index=False)

# ------------------------- 9. Display compact summaries ----------------

print("\n" + "=" * 85)
print("INDIVIDUAL BACKBONE COMPLEXITY SUMMARY")
print("=" * 85)

if not individual_df.empty:
    display(individual_df.round(3))

print("\n" + "=" * 85)
print("COMPLETE AMAFFNET PIPELINE COMPLEXITY SUMMARY")
print("=" * 85)

if not fusion_df.empty:
    display(fusion_df.round(3))

print("\nReports saved:")
print(individual_csv)
print(fusion_csv)

print("\nCOMPUTATIONAL COMPLEXITY ANALYSIS FINISHED.")
print("Review FLOPs and memory fields for missing values before reporting.")
```

```python
# ================================================================
# FLOPs VERIFICATION USING EXPLICIT BATCH SIZE = 1
# ================================================================

import tensorflow as tf
import pandas as pd
import os

from tensorflow.python.framework.convert_to_constants import (
    convert_variables_to_constants_v2
)

BASE = "/content/drive/MyDrive/Microplastics_SEM/AMAFFNet_Weights"

MODEL_PATHS = {
    "Xception_NoSE": f"{BASE}/Xception/best_xception_fold_1.keras",
    "Xception_SE": f"{BASE}/Xception_SE/best_xception_se_fold_1.keras",
    "DenseNet201_NoSE": f"{BASE}/DenseNet201_NoSE/best_densenet201_no_se_fold_1.keras",
    "DenseNet201_SE": f"{BASE}/DenseNet201/best_densenet201_fold_1.keras",
    "MobileNetV2_NoSE": f"{BASE}/MobileNetV2_NoSE/best_mobilenetv2_nose_fold_1.keras",
    "MobileNetV2_SE": f"{BASE}/MobileNetV2/best_mobilenetv2_fold_1.keras"
}

# Previously reported classifier FLOPs
OLD_FLOPS = {
    "Xception_NoSE": 16772000146,
    "Xception_SE": 16773460498,
    "DenseNet201_NoSE": 8633913810,
    "DenseNet201_SE": 8635025610,
    "MobileNetV2_NoSE": 314250962,
    "MobileNetV2_SE": 314725922
}

results = []

for name, path in MODEL_PATHS.items():

    print("\nProfiling:", name)

    # Load the saved classifier checkpoint
    model = tf.keras.models.load_model(path, compile=False)

    # Set batch size explicitly to one
    input_shape = model.input_shape
    concrete_shape = (1,) + tuple(input_shape[1:])

    input_spec = tf.TensorSpec(
        shape=concrete_shape,
        dtype=tf.float32
    )

    # Trace inference with one input image
    traced = tf.function(
        lambda x: model(x, training=False)
    ).get_concrete_function(input_spec)

    # Freeze model variables for TensorFlow graph profiling
    frozen = convert_variables_to_constants_v2(traced)

    # Obtain floating-point operation count
    options = tf.compat.v1.profiler.ProfileOptionBuilder.float_operation()
    options["output"] = "none"

    profile = tf.compat.v1.profiler.profile(
        graph=frozen.graph,
        options=options
    )

    new_flops = int(profile.total_float_ops) if profile else None

    results.append({
        "Model": name,
        "Input_Shape": str(concrete_shape),
        "Old_FLOPs": OLD_FLOPS[name],
        "New_FLOPs_Batch1": new_flops,
        "Difference": (
            new_flops - OLD_FLOPS[name]
            if new_flops is not None else None
        )
    })

    print("Input shape:", concrete_shape)
    print("Previous FLOPs:", OLD_FLOPS[name])
    print("Batch-1 FLOPs:", new_flops)

    del model, traced, frozen

    tf.keras.backend.clear_session()

# Display comparison
df = pd.DataFrame(results)

print("\n" + "=" * 90)
print("FLOPs COMPARISON")
print("=" * 90)

display(df)

# Save verification results
output_path = (
    "/content/drive/MyDrive/Microplastics_SEM/"
    "Computational_Complexity_Results/"
    "FLOPs_Batch1_Verification.csv"
)

os.makedirs(os.path.dirname(output_path), exist_ok=True)
df.to_csv(output_path, index=False)

print("\nSaved:", output_path)
```
